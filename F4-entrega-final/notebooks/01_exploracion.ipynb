```python
# =============================================================================
# 1.1 CLASES DE SOPORTE, TRAZABILIDAD Y PIPELINE MODULAR
# Inspirado en el notebook PV, adaptado al estándar robusto del notebook principal
# =============================================================================

import os
import re
import unicodedata
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
from IPython.display import display, HTML

# -----------------------------------------------------------------------------
# Utilidades transversales
# -----------------------------------------------------------------------------

def normalizar_nombre_columna(col):
    """Normaliza nombres de columnas para uso consistente en todo el notebook."""
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna_por_patron(df_base, patrones):
    """Detecta una columna real del DataFrame usando patrones flexibles."""
    for col in df_base.columns:
        col_norm = normalizar_nombre_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


# -----------------------------------------------------------------------------
# Clase base
# -----------------------------------------------------------------------------

class DatasetBase:
    """
    Clase base para encapsular el DataFrame y registrar transformaciones.
    Su objetivo es entregar trazabilidad metodológica sin modificar la narrativa principal.
    """

    def __init__(self, ruta=None, df=None, nombre="dataset"):
        self.ruta = ruta
        self.df = df.copy(deep=True) if isinstance(df, pd.DataFrame) else df
        self.nombre = nombre
        self.historial = []

    def _registrar(self, paso, descripcion):
        """Registra cada transformación ejecutada sobre el DataFrame."""
        filas = self.df.shape[0] if isinstance(self.df, pd.DataFrame) else 0
        columnas = self.df.shape[1] if isinstance(self.df, pd.DataFrame) else 0
        self.historial.append({
            "paso": paso,
            "descripcion": descripcion,
            "fecha_ejecucion": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "filas": filas,
            "columnas": columnas
        })

    def cargar(self, sep=";", decimal=",", encoding="latin-1"):
        """Carga datos desde CSV o usa un DataFrame ya existente."""
        if isinstance(self.df, pd.DataFrame):
            self._registrar("cargar_memoria", "DataFrame cargado desde memoria")
            return self.df

        if self.ruta is None:
            raise ValueError("Debe indicar una ruta CSV o entregar un DataFrame existente.")

        ruta = Path(self.ruta)
        if not ruta.exists():
            raise FileNotFoundError(f"No se encontró el archivo: {ruta}")

        self.df = pd.read_csv(ruta, sep=sep, decimal=decimal, encoding=encoding)
        self._registrar("cargar_csv", f"Carga desde archivo CSV: {ruta}")
        return self.df

    def normalizar_columnas(self):
        """Normaliza nombres de columnas y registra los cambios."""
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado para normalizar columnas.")

        columnas_antes = list(self.df.columns)
        self.df.columns = [normalizar_nombre_columna(c) for c in self.df.columns]
        cambios = sum(a != b for a, b in zip(columnas_antes, self.df.columns))
        self._registrar("normalizar_columnas", f"Columnas normalizadas: {cambios}")
        return self.df

    def explorar(self):
        """Entrega un resumen estructural mínimo del DataFrame."""
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado para explorar.")

        resumen = {
            "filas": self.df.shape[0],
            "columnas": self.df.shape[1],
            "nulos_totales": int(self.df.isna().sum().sum()),
            "duplicados": int(self.df.duplicated().sum()),
            "memoria_mb": round(self.df.memory_usage(deep=True).sum() / 1024**2, 2)
        }
        self._registrar("explorar", "Resumen estructural del dataset")
        return resumen

    def resumen_historial(self):
        """Retorna el historial de transformaciones como DataFrame."""
        return pd.DataFrame(self.historial)


# -----------------------------------------------------------------------------
# Clase de limpieza
# -----------------------------------------------------------------------------

class DatasetLimpio(DatasetBase):
    """Añade métodos de limpieza reproducibles."""

    def limpiar_fechas(self):
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        col_fecha = detectar_columna_por_patron(self.df, ["fecha", "date"])
        if col_fecha is None:
            self._registrar("limpiar_fechas", "No se detectó columna fecha")
            return self.df

        self.df[col_fecha] = pd.to_datetime(self.df[col_fecha], errors="coerce")
        if pd.api.types.is_datetime64_any_dtype(self.df[col_fecha]):
            self.df[col_fecha] = self.df[col_fecha].dt.floor("h")

        self._registrar("limpiar_fechas", f"Conversión robusta de fecha sobre columna {col_fecha}")
        return self.df

    def eliminar_nulos_criticos(self, columnas=None):
        """Elimina nulos solo en columnas críticas, evitando pérdida excesiva de datos."""
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        if columnas is None:
            columnas = [c for c in [
                detectar_columna_por_patron(self.df, ["fecha"]),
                detectar_columna_por_patron(self.df, ["comuna"]),
                detectar_columna_por_patron(self.df, ["estacion"]),
                detectar_columna_por_patron(self.df, ["mp2_5", "pm2_5", "mp25", "pm25"])
            ] if c is not None]

        antes = len(self.df)
        self.df = self.df.replace(r"^\s*$", np.nan, regex=True)
        self.df = self.df.dropna(subset=columnas).copy()
        despues = len(self.df)

        self._registrar("eliminar_nulos_criticos", f"Eliminadas {antes - despues} filas con nulos críticos")
        return self.df


# -----------------------------------------------------------------------------
# Clase de conversión
# -----------------------------------------------------------------------------

class DatasetConvertido(DatasetLimpio):
    """Añade conversión robusta de booleanos y variables categóricas."""

    def convertir_a_booleanos(self, columnas=None):
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        if columnas is None:
            columnas = ["inversion_termica", "es_finde", "es_festivo"]

        mapa = {
            "sí": 1, "si": 1, "true": 1, "1": 1, "yes": 1, "y": 1,
            "no": 0, "false": 0, "0": 0, "n": 0
        }

        convertidas = []
        for col in columnas:
            if col in self.df.columns:
                self.df[col] = (
                    self.df[col]
                    .astype(str)
                    .str.strip()
                    .str.lower()
                    .map(mapa)
                    .fillna(0)
                    .astype(int)
                )
                convertidas.append(col)

        self._registrar("convertir_booleanos", f"Variables booleanas convertidas: {convertidas}")
        return self.df

    def codificar_one_hot(self, columna, prefijo=None):
        """Codificación one-hot con pandas, evitando dependencias adicionales."""
        if columna not in self.df.columns:
            raise KeyError(f"No existe la columna categórica: {columna}")

        dummies = pd.get_dummies(self.df[columna], prefix=prefijo or columna, dtype=int)
        self.df = pd.concat([self.df.drop(columns=[columna]), dummies], axis=1)
        self._registrar("codificar_one_hot", f"One-hot encoding de {columna}")
        return self.df


# -----------------------------------------------------------------------------
# Clase de filtrado y análisis
# -----------------------------------------------------------------------------

class DatasetFiltrado(DatasetConvertido):
    """Añade criterios de representatividad horaria y agregación diaria."""

    def filtrar_dias_por_horas(self, horas_minimas=18):
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        col_fecha = detectar_columna_por_patron(self.df, ["fecha"])
        col_comuna = detectar_columna_por_patron(self.df, ["comuna"])

        if col_fecha is None or col_comuna is None:
            raise KeyError("No se detectaron columnas fecha y comuna para filtrar por horas.")

        df_aux = self.df.copy()
        df_aux[col_fecha] = pd.to_datetime(df_aux[col_fecha], errors="coerce")
        df_aux = df_aux.dropna(subset=[col_fecha, col_comuna]).copy()
        df_aux["fecha_dia"] = df_aux[col_fecha].dt.date
        df_aux["hora"] = df_aux[col_fecha].dt.hour

        conteo = (
            df_aux.groupby(["fecha_dia", col_comuna])["hora"]
            .nunique()
            .reset_index(name="horas_validas")
        )
        validos = conteo[conteo["horas_validas"] >= horas_minimas]

        antes = len(df_aux)
        df_aux = df_aux.merge(validos[["fecha_dia", col_comuna]], on=["fecha_dia", col_comuna], how="inner")
        df_aux = df_aux.drop(columns=["fecha_dia", "hora"], errors="ignore")
        self.df = df_aux.copy()
        despues = len(self.df)

        self._registrar("filtrar_horas", f"Filtro ≥{horas_minimas} horas: eliminadas {antes - despues} filas")
        return self.df

    def promedios_mp_por_comuna_dia(self):
        """Calcula promedios diarios de MP2.5 y MP10 por comuna, si las columnas existen."""
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        col_fecha = detectar_columna_por_patron(self.df, ["fecha"])
        col_comuna = detectar_columna_por_patron(self.df, ["comuna"])
        col_pm25 = detectar_columna_por_patron(self.df, ["mp2_5", "pm2_5", "mp25", "pm25"])
        col_pm10 = detectar_columna_por_patron(self.df, ["mp10", "pm10"])

        if col_fecha is None or col_comuna is None or col_pm25 is None:
            raise KeyError("Faltan columnas mínimas para calcular promedios diarios.")

        df_aux = self.df.copy()
        df_aux[col_fecha] = pd.to_datetime(df_aux[col_fecha], errors="coerce")
        df_aux["fecha_dia"] = df_aux[col_fecha].dt.date
        df_aux[col_pm25] = pd.to_numeric(df_aux[col_pm25], errors="coerce")

        agg = {col_pm25: "mean"}
        if col_pm10 is not None:
            df_aux[col_pm10] = pd.to_numeric(df_aux[col_pm10], errors="coerce")
            agg[col_pm10] = "mean"

        resultado = df_aux.groupby(["fecha_dia", col_comuna], as_index=False).agg(agg)
        rename = {col_pm25: "mp2_5_promedio_diario"}
        if col_pm10 is not None:
            rename[col_pm10] = "mp10_promedio_diario"
        resultado = resultado.rename(columns=rename)

        self._registrar("promedios_diarios", "Cálculo de promedios diarios por comuna")
        return resultado

    def contar_dias_sobre_umbral(self, df_promedios, umbral_pm25=49, umbral_pm10=179):
        """Cuenta eventos sobre umbral en un DataFrame de promedios diarios."""
        resultado = {
            "total_registros": len(df_promedios),
            "umbral_pm25": umbral_pm25,
            "umbral_pm10": umbral_pm10
        }

        if "mp2_5_promedio_diario" in df_promedios.columns:
            resultado["eventos_pm25"] = int((df_promedios["mp2_5_promedio_diario"] > umbral_pm25).sum())
        if "mp10_promedio_diario" in df_promedios.columns:
            resultado["eventos_pm10"] = int((df_promedios["mp10_promedio_diario"] > umbral_pm10).sum())

        self._registrar("contar_umbral", "Conteo de episodios críticos sobre umbral")
        return resultado

    def validar(self):
        """Validación general para trazabilidad técnica."""
        if not isinstance(self.df, pd.DataFrame):
            raise ValueError("No existe DataFrame cargado.")

        reporte = {
            "filas": self.df.shape[0],
            "columnas": self.df.shape[1],
            "nulos_totales": int(self.df.isna().sum().sum()),
            "duplicados": int(self.df.duplicated().sum()),
            "columnas_texto": self.df.select_dtypes(exclude=[np.number, "datetime", "datetimetz", "bool"]).columns.tolist()
        }
        self._registrar("validar", "Validación técnica general del dataset")
        return reporte


# -----------------------------------------------------------------------------
# Función de exportación trazable
# -----------------------------------------------------------------------------

def guardar_dataset_trazable(df, nombre_archivo="sinca_limpio.csv", carpeta="../data/processed"):
    """Guarda un DataFrame y retorna la ruta de salida."""
    carpeta = Path(carpeta)
    carpeta.mkdir(parents=True, exist_ok=True)
    ruta_salida = carpeta / nombre_archivo
    df.to_csv(ruta_salida, index=False, encoding="utf-8")
    return ruta_salida

print("✓ Capa POO de trazabilidad cargada correctamente.")
print("✓ Clases disponibles: DatasetBase, DatasetLimpio, DatasetConvertido, DatasetFiltrado.")
```

```python

# =============================================================================
# 3. CARGA, NORMALIZACIÓN INICIAL Y ENRIQUECIMIENTO DEL DATASET
# Proyecto F4 (Entrega final): Calidad del Aire en Santiago
# Fuente: SINCA
# =============================================================================

from pathlib import Path
from datetime import datetime
import pandas as pd
import numpy as np

# -----------------------------------------------------------------------------
# 1) Configuración del archivo fuente
# -----------------------------------------------------------------------------

ruta = Path("../data/raw/sinca_santiago.csv")

#print("╔════════════════════════════════════════════════════════════════════╗")
#print("║        CARGA, NORMALIZACIÓN Y ENRIQUECIMIENTO DEL DATASET         ║")
#print("╚════════════════════════════════════════════════════════════════════╝")
#print(f"Fecha de ejecución : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
#print(f"Archivo fuente     : {ruta}")
#print(f"Archivo existe     : {'✓ Sí' if ruta.exists() else '✗ No'}")

if not ruta.exists():
    raise FileNotFoundError(f"No se encontró el archivo fuente: {ruta.resolve()}")

# -----------------------------------------------------------------------------
# 2) Carga robusta del CSV SINCA
# -----------------------------------------------------------------------------

df = pd.read_csv(
    ruta,
    sep=";",
    decimal=",",
    encoding="latin-1"
)

filas_originales, columnas_originales = df.shape

# -----------------------------------------------------------------------------
# 3) Limpieza estructural básica
# -----------------------------------------------------------------------------

filas_vacias = int(df.isna().all(axis=1).sum())
df = df.dropna(how="all").copy()

# -----------------------------------------------------------------------------
# 4) Conversión robusta de fecha
# -----------------------------------------------------------------------------

if "fecha" not in df.columns:
    raise KeyError("La columna 'fecha' no existe en el dataset.")

estrategia_fecha = "pd.to_datetime(format='mixed', cache=True)"

try:
    df["fecha"] = pd.to_datetime(df["fecha"], format="mixed", cache=True)
except Exception as e:
    estrategia_fecha = "pd.to_datetime(errors='coerce')"
    print(f"⚠ Aviso: conversión automática de fecha falló ({e}). Aplicando plan alternativo.")
    df["fecha"] = pd.to_datetime(df["fecha"], errors="coerce")

fechas_invalidas = int(df["fecha"].isna().sum())
df["fecha"] = df["fecha"].dt.floor("h")

# -----------------------------------------------------------------------------
# 5) Diccionario semántico de variables y unidades
# -----------------------------------------------------------------------------

dicc_columnas = {
    "MP2.5": ["µg/m³", "cuantitativo"],
    "MP10": ["µg/m³", "cuantitativo"],
    "temperatura": ["°C", "cuantitativo"],
    "humedad": ["%", "cuantitativo"],
    "presion": ["hPa", "cuantitativo"],
    "viento": ["m/s", "cuantitativo"],
    "radiacion": ["W/m²", "cuantitativo"],
    "inversion_termica": ["", "cualitativo"],
    "dia_semana": ["", "cualitativo"],
    "es_finde": ["", "cualitativo"],
    "es_festivo": ["", "cualitativo"],
    "fecha": ["", "temporal"],
    "estacion": ["", "cualitativo"],
    "comuna": ["", "cualitativo"],
}

dicc_dias_semana = {
    0: "Lunes",
    1: "Martes",
    2: "Miércoles",
    3: "Jueves",
    4: "Viernes",
    5: "Sábado",
    6: "Domingo",
}

dicc_booleano = {
    0: "No",
    1: "Sí",
}

mapeos = {
    "dia_semana": dicc_dias_semana,
    "es_finde": dicc_booleano,
    "es_festivo": dicc_booleano,
    "inversion_termica": dicc_booleano,
}

# -----------------------------------------------------------------------------
# 6) Homologación de variables categóricas
# -----------------------------------------------------------------------------

columnas_mapeadas = []
columnas_no_encontradas_mapeo = []

for col, dicc in mapeos.items():
    if col in df.columns:
        valores_antes = df[col].dropna().nunique()
        df[col] = df[col].map(dicc).fillna(df[col])
        valores_despues = df[col].dropna().nunique()
        columnas_mapeadas.append(f"{col} ({valores_antes}→{valores_despues})")
    else:
        columnas_no_encontradas_mapeo.append(col)

# -----------------------------------------------------------------------------
# 7) Enriquecimiento de encabezados con unidades físicas
# -----------------------------------------------------------------------------

nuevos_encabezados = {
    col: f"{col} ({valores[0]})" if valores[0].strip() else col
    for col, valores in dicc_columnas.items()
    if col in df.columns
}

df.rename(columns=nuevos_encabezados, inplace=True)

# -----------------------------------------------------------------------------
# 8) Clasificación analítica de variables
# -----------------------------------------------------------------------------

variables_cuantitativas = [
    nuevos_encabezados.get(col, col)
    for col, valores in dicc_columnas.items()
    if valores[1] == "cuantitativo" and nuevos_encabezados.get(col, col) in df.columns
]

variables_cualitativas = [
    nuevos_encabezados.get(col, col)
    for col, valores in dicc_columnas.items()
    if valores[1] == "cualitativo" and nuevos_encabezados.get(col, col) in df.columns
]

variables_temporales = [
    nuevos_encabezados.get(col, col)
    for col, valores in dicc_columnas.items()
    if valores[1] == "temporal" and nuevos_encabezados.get(col, col) in df.columns
]

# -----------------------------------------------------------------------------
# 9) Controles de calidad inicial
# -----------------------------------------------------------------------------

duplicados = int(df.duplicated().sum())
nulos_totales = int(df.isna().sum().sum())

fecha_min = df["fecha"].min()
fecha_max = df["fecha"].max()

columnas_con_unidad = [
    col for col in df.columns
    if "(" in col and ")" in col
]

# Versión formateada solo para visualización
df_preview = df.copy()
df_preview["fecha"] = pd.to_datetime(df_preview["fecha"], errors="coerce").dt.strftime("%Y-%m-%d %I:%M %p")

# -----------------------------------------------------------------------------
# 10) Output ejecutivo, robusto y no truncable
# -----------------------------------------------------------------------------

from IPython.display import display, HTML, Markdown
resumen_html = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:18px;
    border-left:6px solid #1F3A5F;
    border-radius:10px;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;">
✅ 3. Resultado de Carga y Normalización Inicial
</h3>
<table style="
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    color:#111827;
">
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Archivo fuente</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{ruta.name}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas originales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{filas_originales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Columnas originales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{columnas_originales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas vacías eliminadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{filas_vacias:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{df.shape[0]:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Columnas finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{df.shape[1]:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Fechas inválidas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{fechas_invalidas:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros duplicados</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{duplicados:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Valores nulos totales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{nulos_totales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Memoria utilizada</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{df.memory_usage(deep=True).sum():,} bytes</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Fecha mínima</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{fecha_min}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Fecha máxima</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{fecha_max}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;"><strong>Granularidad temporal</strong></td><td style="text-align:left;padding:4px 8px;">Horaria</td></tr>
</table>
<p style="
    margin-top:14px;
    margin-bottom:0;
    font-size:14px;
    line-height:1.7;
    color:#111827;
">
<strong>Conclusión técnica:</strong> El dataset fue cargado correctamente, respetando el formato SINCA
(<strong>sep=';'</strong>, <strong>decimal=','</strong>, <strong>encoding='latin-1'</strong>).
La variable temporal fue normalizada a granularidad horaria y las variables fueron enriquecidas
con unidades físicas y clasificación analítica.
</p>
</div>
"""
display(HTML(resumen_html))

# Vista controlada del dataset: evita truncamiento excesivo
columnas_preview = list(df_preview.columns[:8])

print("Vista preliminar controlada del dataset:")
print(f"Mostrando primeras 5 filas y {len(columnas_preview)} columnas de {df_preview.shape[1]} columnas totales.")

display(df_preview[columnas_preview].head(5))
```

```python
# =============================================================================
# 5. LIMPIEZA Y PREPARACIÓN DEL DATASET
# =============================================================================

import re
import unicodedata
import pandas as pd
from IPython.display import display, HTML, Markdown

# -----------------------------------------------------------------------------
# Validación inicial
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "El DataFrame principal 'df' no existe. "
        "Ejecute primero la etapa de carga de datos."
    )

df_limpio = df.copy()

# -----------------------------------------------------------------------------
# Diagnóstico previo
# -----------------------------------------------------------------------------

registros_iniciales = df_limpio.shape[0]
variables_iniciales = df_limpio.shape[1]

nulos_iniciales = int(df_limpio.isna().sum().sum())
duplicados_iniciales = int(df_limpio.duplicated().sum())

# Ranking controlado de nulos
top_nulos = (
    df_limpio.isna()
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

top_nulos.columns = ["Variable", "Valores nulos"]
top_nulos = top_nulos[top_nulos["Valores nulos"] > 0].head(10)

# -----------------------------------------------------------------------------
# Limpieza controlada e Inyección de Regla de Negocio (Horas Mínimas)
# -----------------------------------------------------------------------------

# 1. Identificación robusta de columnas temporales y espaciales
def _normalizar_texto(txt):
    txt = str(txt).strip().lower()
    txt = unicodedata.normalize("NFKD", txt)
    return "".join(c for c in txt if not unicodedata.combining(c))

col_fecha = next((c for c in df_limpio.columns if "fech" in _normalizar_texto(c) or "date" in _normalizar_texto(c)), None)
col_comuna = next((c for c in df_limpio.columns if "comun" in _normalizar_texto(c) or "station" in _normalizar_texto(c)), None)

# 2. Aplicación del filtro si se encuentran las columnas requeridas
if col_fecha and col_comuna:
    # Asegurar tipo datetime para extraer solo el componente del día (fecha_id)
    df_limpio[col_fecha] = pd.to_datetime(df_limpio[col_fecha])
    df_limpio['_fecha_dia'] = df_limpio[col_fecha].dt.date
    
    # Calcular cuántas horas registradas tiene cada combinación Comuna-Día
    # (Cada fila representa una hora en estructuras de datos de calidad del aire estándar)
    conteo_horas = df_limpio.groupby([col_comuna, '_fecha_dia'])[col_fecha].transform('count')
    
    # Filtrar preservando únicamente los días con 18 o más mediciones horarias
    df_limpio = df_limpio[conteo_horas >= 18].copy()
    
    # Eliminar columna auxiliar
    df_limpio = df_limpio.drop(columns=['_fecha_dia'])
else:
    print("⚠️ Advertencia: No se detectó columna de 'fecha' o 'comuna'. Saltando filtro de 18 horas.")

# 3. Eliminación de filas con valores nulos (Comportamiento original del script)
df_limpio = (
    df_limpio
    .dropna()
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# Métricas posteriores
# -----------------------------------------------------------------------------

registros_finales = df_limpio.shape[0]

registros_eliminados = registros_iniciales - registros_finales

porcentaje_eliminado = (
    registros_eliminados / registros_iniciales * 100
    if registros_iniciales > 0
    else 0
)

nulos_finales = int(df_limpio.isna().sum().sum())
duplicados_finales = int(df_limpio.duplicated().sum())

# -----------------------------------------------------------------------------
# Resumen ejecutivo
# -----------------------------------------------------------------------------

html_resumen = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:18px;
    border-left:6px solid #B22234;
    border-radius:10px;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;">
🧹 5. Resultado de la Limpieza del Dataset
</h3>
<table style="
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    color:#111827;
">
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros iniciales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{registros_iniciales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{registros_finales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros eliminados</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{registros_eliminados:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>% eliminado</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{porcentaje_eliminado:.2f}%</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Nulos iniciales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{nulos_iniciales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Nulos finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{nulos_finales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Duplicados iniciales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{duplicados_iniciales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;"><strong>Duplicados finales</strong></td><td style="text-align:left;padding:4px 8px;">{duplicados_finales:,}</td></tr>
</table>
<p style="margin-top:15px;margin-bottom:0;font-size:14px;line-height:1.6;color:#111827;">
<strong>Conclusión:</strong> El proceso de limpieza eliminó los registros incompletos identificados durante la etapa de diagnóstico y aplicó un filtro de representatividad temporal (mínimo 18 horas por comuna/día), obteniendo un conjunto de datos consistente y preparado para análisis exploratorio y modelamiento posterior.
</p>
</div>
"""
display(HTML(html_resumen))

# -----------------------------------------------------------------------------
# Evidencia de calidad
# -----------------------------------------------------------------------------

if not top_nulos.empty:
    print("Variables con mayor cantidad de valores faltantes antes de la limpieza:")
    display(top_nulos)

print("\nVista preliminar del dataset limpio:")
display(df_limpio.head())

# Actualización del dataset principal
df = df_limpio.copy()
```

```python
# =============================================================================
# 6. NORMALIZACIÓN FINAL Y EXPORTACIÓN DEL DATASET PROCESADO
# Proyecto F4 (Entrega final): Calidad del Aire en Santiago
# =============================================================================

import re
import unicodedata
from pathlib import Path
from datetime import datetime

from IPython.display import display, HTML, Markdown

# -----------------------------------------------------------------------------
# Validación inicial
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "El DataFrame principal 'df' no existe. "
        "Ejecute primero las etapas de carga y limpieza del dataset."
    )

df_export = df.copy()

# -----------------------------------------------------------------------------
# Función robusta de normalización de columnas
# -----------------------------------------------------------------------------

def normalizar_columna(columna: str) -> str:
    """
    Normaliza nombres de columnas para uso programático:
    - Elimina tildes.
    - Convierte a minúsculas.
    - Reemplaza espacios y símbolos por guion bajo.
    - Elimina duplicidad de guiones bajos.
    """
    col = str(columna).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")

columnas_originales = list(df_export.columns)
df_export.columns = [normalizar_columna(c) for c in df_export.columns]
columnas_normalizadas = list(df_export.columns)

# -----------------------------------------------------------------------------
# Validación de nombres duplicados luego de normalización
# -----------------------------------------------------------------------------

duplicadas = pd.Series(columnas_normalizadas).duplicated()
if duplicadas.any():
    columnas_dup = pd.Series(columnas_normalizadas)[duplicadas].tolist()
    raise ValueError(
        f"Se detectaron columnas duplicadas luego de la normalización: {columnas_dup}"
    )

# -----------------------------------------------------------------------------
# Definición de variables clave para limpieza analítica final
# -----------------------------------------------------------------------------

columnas_clave = [
    "mp2_5",
    "mp10",
    "temperatura",
    "humedad",
    "presion",
    "viento",
    "radiacion",
]

columnas_existentes = [c for c in columnas_clave if c in df_export.columns]
columnas_faltantes = [c for c in columnas_clave if c not in df_export.columns]

if not columnas_existentes:
    raise ValueError(
        "No se encontraron columnas clave para limpieza final. "
        f"Columnas disponibles: {df_export.columns.tolist()[:20]}"
    )

# -----------------------------------------------------------------------------
# Limpieza final sobre columnas relevantes
# -----------------------------------------------------------------------------

filas_antes = df_export.shape[0]
nulos_clave_antes = int(df_export[columnas_existentes].isna().sum().sum())

df_export = (
    df_export
    .dropna(subset=columnas_existentes)
    .drop_duplicates()
    .reset_index(drop=True)
)

filas_despues = df_export.shape[0]
filas_eliminadas = filas_antes - filas_despues
porcentaje_eliminado = (
    filas_eliminadas / filas_antes * 100
    if filas_antes > 0
    else 0
)

nulos_clave_despues = int(df_export[columnas_existentes].isna().sum().sum())
duplicados_despues = int(df_export.duplicated().sum())

# -----------------------------------------------------------------------------
# Exportación del dataset procesado
# -----------------------------------------------------------------------------

ruta_salida = Path("../data/processed/sinca_limpio.csv")
ruta_salida.parent.mkdir(parents=True, exist_ok=True)

df_export.to_csv(
    ruta_salida,
    index=False,
    encoding="utf-8"
)

archivo_generado = ruta_salida.exists()
tamano_mb = ruta_salida.stat().st_size / 1024**2 if archivo_generado else 0

# Actualizar DataFrame principal
df = df_export.copy()

# -----------------------------------------------------------------------------
# Evidencia controlada de normalización de columnas
# -----------------------------------------------------------------------------

cambios_columnas = pd.DataFrame({
    "Columna original": columnas_originales,
    "Columna normalizada": columnas_normalizadas
})

cambios_columnas = cambios_columnas[
    cambios_columnas["Columna original"] != cambios_columnas["Columna normalizada"]
].head(12)

# -----------------------------------------------------------------------------
# Output ejecutivo no truncable
# -----------------------------------------------------------------------------

html_exportacion = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:18px;
    border-left:6px solid #1F3A5F;
    border-radius:10px;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;">
💾 6. Resultado de Normalización y Exportación
</h3>
<table style="
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    color:#111827;
">
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Fecha de ejecución</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas antes de limpieza final</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{filas_antes:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas finales exportadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{filas_despues:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Filas eliminadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{filas_eliminadas:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>% eliminado</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{porcentaje_eliminado:.2f}%</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Columnas finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{df_export.shape[1]:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Columnas clave usadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{', '.join(columnas_existentes)}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Columnas clave faltantes</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{', '.join(columnas_faltantes) if columnas_faltantes else 'Ninguna'}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Nulos en columnas clave antes</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{nulos_clave_antes:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Nulos en columnas clave después</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{nulos_clave_despues:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Duplicados finales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{duplicados_despues:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Archivo generado</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{'✔ Sí' if archivo_generado else '✘ No'}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Ruta salida</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{ruta_salida}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;"><strong>Tamaño archivo</strong></td><td style="text-align:left;padding:4px 8px;">{tamano_mb:.2f} MB</td></tr>
</table>
<p style="
    margin-top:14px;
    margin-bottom:0;
    font-size:14px;
    line-height:1.7;
    color:#111827;
">
<strong>Conclusión técnica:</strong> El dataset fue normalizado, depurado en sus variables analíticas clave y exportado correctamente como archivo procesado, manteniendo separación metodológica entre datos crudos y datos preparados.
</p>
</div>
"""
display(HTML(html_exportacion))

if not cambios_columnas.empty:
    print("Ejemplo controlado de columnas normalizadas:")
    display(cambios_columnas)

print("Vista preliminar controlada del dataset procesado:")
display(df.head(5))
```

```python
# =============================================================================
# 7. DEFINICIÓN Y CONSTRUCCIÓN DE EPISODIOS CRÍTICOS (CON METEOROLOGÍA POR COMUNA)
# =============================================================================

import pandas as pd
import numpy as np
from IPython.display import display, HTML

# -----------------------------------------------------------------------------
# Validaciones iniciales
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "El DataFrame principal 'df' no existe. "
        "Ejecute primero las etapas anteriores."
    )

df_eventos = df.copy()

columnas_requeridas = ["fecha", "comuna"]
faltantes = [c for c in columnas_requeridas if c not in df_eventos.columns]

if faltantes:
    raise KeyError(
        "Faltan columnas obligatorias: " f"{faltantes}. "
        f"Columnas disponibles: {df_eventos.columns.tolist()}"
    )

# -----------------------------------------------------------------------------
# Detección robusta de columnas de contaminantes
# -----------------------------------------------------------------------------

def detectar_columna(df_base, patrones):
    for col in df_base.columns:
        col_norm = col.lower().replace(".", "").replace("_", "").replace(" ", "")
        if any(p in col_norm for p in patrones):
            return col
    return None

col_pm25 = detectar_columna(df_eventos, ["mp25", "pm25", "mp2,5", "pm2,5", "mp2"])
col_pm10 = detectar_columna(df_eventos, ["mp10", "pm10"])

print("Columna PM2.5 detectada:", col_pm25)
print("Columna PM10 detectada:", col_pm10)

if col_pm25 is None:
    raise KeyError(
        "No se detectó columna para PM2.5 / MP2.5. "
        f"Columnas disponibles: {df_eventos.columns.tolist()}"
    )

# MP10 queda como opcional para evitar errores si no existe
contaminantes_detectados = [col_pm25]
if col_pm10 is not None:
    contaminantes_detectados.append(col_pm10)

# -----------------------------------------------------------------------------
# Normalización robusta de fecha
# -----------------------------------------------------------------------------

df_eventos["fecha"] = pd.to_datetime(
    df_eventos["fecha"],
    errors="coerce",
    dayfirst=False
)

fechas_invalidas = df_eventos["fecha"].isna().sum()

if fechas_invalidas > 0:
    print(f"Advertencia: se eliminaron {fechas_invalidas:,} registros con fecha inválida.")

df_eventos = df_eventos.dropna(subset=["fecha", "comuna"]).copy()

# -----------------------------------------------------------------------------
# Construcción de fecha diaria
# -----------------------------------------------------------------------------

df_eventos["fecha_dia"] = df_eventos["fecha"].dt.date

# -----------------------------------------------------------------------------
# Filtrado normativo: mínimo 18 observaciones diarias por comuna
# -----------------------------------------------------------------------------

registros_antes = len(df_eventos)

df_eventos = (
    df_eventos
    .groupby(["fecha_dia", "comuna"], group_keys=False)
    .filter(lambda grupo: len(grupo) >= 18)
    .copy()
)

registros_despues = len(df_eventos)
registros_descartados = registros_antes - registros_despues

if registros_despues == 0:
    raise ValueError(
        "Después del filtro de mínimo 18 observaciones diarias no quedaron registros válidos. "
        "Revise frecuencia temporal, columna fecha o granularidad por comuna."
    )

# -----------------------------------------------------------------------------
# Promedios diarios y Extracción de Métricas Meteorológicas (Agrupado por Comuna)
# -----------------------------------------------------------------------------

# Identificar dinámicamente el resto de las variables numéricas (temperatura, humedad, etc.)
excluir_columnas = ["fecha", "comuna", "fecha_dia"] + contaminantes_detectados
columnas_meteorologicas = [
    col for col in df_eventos.columns 
    if col not in excluir_columnas and pd.api.types.is_numeric_dtype(df_eventos[col])
]

# Construir diccionario de agregación mapeando contaminantes y variables del clima
agg_dict = {}

# Los contaminantes conservan estrictamente su promedio diario (lógica original)
for col in contaminantes_detectados:
    agg_dict[col] = ["mean"]

# Las variables meteorológicas adoptan las tres métricas solicitadas
for col in columnas_meteorologicas:
    agg_dict[col] = ["mean", "min", "max"]

# Ejecutar agregación agrupada estrictamente por día y comuna
promedios_diarios = (
    df_eventos
    .groupby(["fecha_dia", "comuna"], as_index=False)
    .agg(agg_dict)
)

# SOLUCIÓN AL APRECIADO ERROR: Aplanar de forma segura respetando las agrupaciones
nuevas_columnas = []
for col, metrica in promedios_diarios.columns:
    if col in ["fecha_dia", "comuna"]:
        nuevas_columnas.append(col) # Preserva las columnas llave de la agrupación
    elif col in contaminantes_detectados:
        nuevas_columnas.append(col) # Deja el nombre original para no romper lógica de umbrales
    else:
        # Renombrado por comuna/día: ej. temperatura_promedio, temperatura_min, temperatura_max
         sufijo = "promedio" if metrica == "mean" else metrica
         nuevas_columnas.append(f"{col}_{sufijo}")

promedios_diarios.columns = nuevas_columnas

# -----------------------------------------------------------------------------
# Construcción de variables objetivo
# -----------------------------------------------------------------------------

UMBRAL_PM25 = 49
UMBRAL_PM10 = 179

promedios_diarios["evento_pm25"] = (
    promedios_diarios[col_pm25] > UMBRAL_PM25
).astype(int)

if col_pm10 is not None:
    promedios_diarios["evento_pm10"] = (
        promedios_diarios[col_pm10] > UMBRAL_PM10
    ).astype(int)
else:
    promedios_diarios["evento_pm10"] = 0

# -----------------------------------------------------------------------------
# Renombrar columnas a nombres estándar
# -----------------------------------------------------------------------------

rename_dict = {
    col_pm25: "mp2_5_promedio_diario"
}

if col_pm10 is not None:
    rename_dict[col_pm10] = "mp10_promedio_diario"
else:
    promedios_diarios["mp10_promedio_diario"] = np.nan

promedios_diarios = promedios_diarios.rename(columns=rename_dict)

# -----------------------------------------------------------------------------
# Métricas ejecutivas
# -----------------------------------------------------------------------------

episodios_pm25 = int(promedios_diarios["evento_pm25"].sum())
episodios_pm10 = int(promedios_diarios["evento_pm10"].sum())
dias_analizados = len(promedios_diarios)

texto_conclusion = (
    "Se construyó la variable objetivo de episodios críticos para PM2.5 y PM10 "
    "según umbrales definidos y validación mínima de 18 observaciones diarias."
)

if episodios_pm10 == 0:
    texto_conclusion = (
        "Considerando que MP10 no tuvo episodios críticos detectados, "
        "el análisis posterior se concentrará principalmente en MP2.5."
    )

# -----------------------------------------------------------------------------
# Resumen ejecutivo HTML
# -----------------------------------------------------------------------------

html_eventos = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:20px;
    border-left:6px solid #B22234;
    border-radius:10px;
    font-family:Arial,Helvetica,sans-serif;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;text-align:left;">
🚨 7. Construcción de Episodios Críticos
</h3>
<table style="
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    color:#111827;
    text-align:left;
">
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros originales</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_antes:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros válidos</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_despues:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros descartados</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_descartados:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Días / comunas analizados</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{dias_analizados:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Episodios PM2.5</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{episodios_pm25:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Episodios PM10</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{episodios_pm10:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Umbral PM2.5</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">49 µg/m³</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Umbral PM10</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">179 µg/m³</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Mínimo observaciones</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">18 horas</td></tr>
</table>
<p style="
    margin-top:16px;
    margin-bottom:0;
    line-height:1.5;
    font-size:14px;
    color:#111827;
    text-align:left;
">
<strong>Conclusión técnica:</strong> {texto_conclusion}
</p>
</div>
"""
display(HTML(html_eventos))

# -----------------------------------------------------------------------------
# Evidencia controlada (Visualización extendida por Comuna)
# -----------------------------------------------------------------------------

print("Vista preliminar de episodios críticos construidos con variables meteorológicas por comuna:")

# Construir las columnas para el print asegurando el orden correcto
columnas_preview = [
    "fecha_dia",
    "comuna",
    "mp2_5_promedio_diario",
    "mp10_promedio_diario"
]

for col in columnas_meteorologicas:
    columnas_preview.extend([f"{col}_promedio", f"{col}_min", f"{col}_max"])

columnas_preview.extend(["evento_pm25", "evento_pm10"])

# Mostrar las columnas deseadas agrupadas correctamente
display(promedios_diarios[columnas_preview].head(10))

# DataFrame final para las siguientes etapas
df_eventos_criticos = promedios_diarios.copy()

print("DataFrame final creado correctamente: df_eventos_criticos")
```

```python
# =============================================================================
# 7.1. CÁLCULO DE PROMEDIOS DIARIOS Y DETECCIÓN DE EPISODIOS CRÍTICOS
# Proyecto F4 (Entrega final): Calidad del Aire en Santiago
# =============================================================================

import re
import unicodedata
import pandas as pd
from IPython.display import display, HTML, Markdown

# -----------------------------------------------------------------------------
# 1) Validación del DataFrame base
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "No existe el DataFrame principal 'df'. "
        "Ejecute primero las etapas de carga, limpieza y exportación."
    )

df_work = df.copy()

# -----------------------------------------------------------------------------
# 2) Normalización segura de nombres de columnas
# -----------------------------------------------------------------------------

def normalizar_columna(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")

df_work.columns = [normalizar_columna(c) for c in df_work.columns]

# -----------------------------------------------------------------------------
# 3) Validación flexible de columnas requeridas
# -----------------------------------------------------------------------------

def buscar_columna(columnas, patrones):
    for col in columnas:
        col_norm = normalizar_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None

col_fecha = buscar_columna(df_work.columns, ["fecha"])
col_comuna = buscar_columna(df_work.columns, ["comuna"])
col_estacion = buscar_columna(df_work.columns, ["estacion"])
col_pm25 = buscar_columna(df_work.columns, ["mp2_5", "pm2_5", "mp25", "pm25"])
col_pm10 = buscar_columna(df_work.columns, ["mp10", "pm10"])

columnas_detectadas = {
    "fecha": col_fecha,
    "comuna": col_comuna,
    "estacion": col_estacion,
    "mp2_5": col_pm25
}

faltantes = [k for k, v in columnas_detectadas.items() if v is None]

if faltantes:
    raise KeyError(
        f"Faltan columnas requeridas: {faltantes}. "
        f"Columnas disponibles: {df_work.columns.tolist()}"
    )

df_work = df_work.rename(columns={
    col_fecha: "fecha",
    col_comuna: "comuna",
    col_estacion: "estacion",
    col_pm25: "mp2_5"
})

if col_pm10 is not None:
    df_work = df_work.rename(columns={col_pm10: "mp10"})
    tiene_pm10 = True
else:
    tiene_pm10 = False

print("Columnas detectadas correctamente:")
print(f"Fecha    : {col_fecha}")
print(f"Comuna   : {col_comuna}")
print(f"Estación : {col_estacion}")
print(f"PM2.5    : {col_pm25}")
print(f"PM10     : {col_pm10 if col_pm10 else 'No detectada'}")

# -----------------------------------------------------------------------------
# 4) Conversión robusta de fecha y contaminantes
# -----------------------------------------------------------------------------

df_work["fecha"] = pd.to_datetime(df_work["fecha"], errors="coerce")
df_work = df_work.dropna(subset=["fecha"]).copy()

df_work["mp2_5"] = pd.to_numeric(df_work["mp2_5"], errors="coerce")

if tiene_pm10:
    df_work["mp10"] = pd.to_numeric(df_work["mp10"], errors="coerce")

df_work["solo_fecha"] = df_work["fecha"].dt.date

# -----------------------------------------------------------------------------
# 4) Conversión robusta de fecha y contaminantes
# -----------------------------------------------------------------------------

df_work["fecha"] = pd.to_datetime(df_work["fecha"], errors="coerce")
df_work = df_work.dropna(subset=["fecha"]).copy()

df_work["mp2_5"] = pd.to_numeric(df_work["mp2_5"], errors="coerce")

if tiene_pm10:
    df_work["mp10"] = pd.to_numeric(df_work["mp10"], errors="coerce")

df_work["solo_fecha"] = df_work["fecha"].dt.date

# -----------------------------------------------------------------------------
# 5) Validación normativa: mínimo 18 registros horarios por día/comuna/estación
# -----------------------------------------------------------------------------

conteo_horario = (
    df_work
    .groupby(["solo_fecha", "comuna", "estacion"], as_index=False)
    .size()
    .rename(columns={"size": "horas_validas"})
)

grupos_validos = conteo_horario[conteo_horario["horas_validas"] >= 18]

df_work = df_work.merge(
    grupos_validos[["solo_fecha", "comuna", "estacion"]],
    on=["solo_fecha", "comuna", "estacion"],
    how="inner"
)

# -----------------------------------------------------------------------------
# 6) Promedios diarios
# -----------------------------------------------------------------------------

agregaciones = {
    "mp2_5": "mean"
}

if tiene_pm10:
    agregaciones["mp10"] = "mean"

promedios_diarios = (
    df_work
    .groupby(["solo_fecha", "comuna", "estacion"], as_index=False)
    .agg(agregaciones)
)

promedios_diarios = promedios_diarios.rename(columns={
    "mp2_5": "mp2_5_promedio_diario",
    "mp10": "mp10_promedio_diario"
})

# -----------------------------------------------------------------------------
# 7) Umbrales normativos y detección de episodios críticos
# -----------------------------------------------------------------------------

UMBRAL_PM25 = 49
UMBRAL_PM10 = 179

promedios_diarios["episodio_pm25"] = (
    promedios_diarios["mp2_5_promedio_diario"] > UMBRAL_PM25
)

if tiene_pm10 and "mp10_promedio_diario" in promedios_diarios.columns:
    promedios_diarios["episodio_pm10"] = (
        promedios_diarios["mp10_promedio_diario"] > UMBRAL_PM10
    )
else:
    promedios_diarios["episodio_pm10"] = False

mp25_critico = (
    promedios_diarios[promedios_diarios["episodio_pm25"]]
    .sort_values(by=["solo_fecha", "comuna", "estacion"])
    .reset_index(drop=True)
)

mp10_critico = (
    promedios_diarios[promedios_diarios["episodio_pm10"]]
    .sort_values(by=["solo_fecha", "comuna", "estacion"])
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 8) Métricas ejecutivas
# -----------------------------------------------------------------------------

registros_originales = len(df)
registros_validos = len(df_work)
combinaciones_analizadas = len(promedios_diarios)
dias_validos = promedios_diarios["solo_fecha"].nunique()
estaciones_validas = promedios_diarios["estacion"].nunique()
comunas_validas = promedios_diarios["comuna"].nunique()

episodios_pm25 = len(mp25_critico)
episodios_pm10 = len(mp10_critico)

porcentaje_pm25 = (
    episodios_pm25 / combinaciones_analizadas * 100
    if combinaciones_analizadas > 0
    else 0
)

porcentaje_pm10 = (
    episodios_pm10 / combinaciones_analizadas * 100
    if combinaciones_analizadas > 0
    else 0
)

# -----------------------------------------------------------------------------
# 9) Output ejecutivo profesional
# -----------------------------------------------------------------------------

html_eventos = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:18px;
    border-left:6px solid #B22234;
    border-radius:10px;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;">
🚨 7.1 Resultado de Detección de Episodios Críticos
</h3>
<table style="
    width:100%;
    border-collapse:collapse;
    font-size:14px;
    color:#111827;
">
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros originales</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{registros_originales:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Registros válidos tras filtro ≥18h</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{registros_validos:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Combinaciones día/comuna/estación analizadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{combinaciones_analizadas:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Días válidos analizados</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{dias_validos:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Comunas analizadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{comunas_validas:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Estaciones analizadas</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{estaciones_validas:,}</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Umbral PM2.5</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{UMBRAL_PM25} µg/m³</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Umbral PM10</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{UMBRAL_PM10} µg/m³</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;"><strong>Episodios PM2.5 detectados</strong></td><td style="text-align:left;padding:4px 8px;border-bottom:1px solid #cbd5e1;">{episodios_pm25:,} ({porcentaje_pm25:.2f}%)</td></tr>
  <tr><td style="text-align:left;padding:4px 8px;"><strong>Episodios PM10 detectados</strong></td><td style="text-align:left;padding:4px 8px;">{episodios_pm10:,} ({porcentaje_pm10:.2f}%)</td></tr>
</table>
<p style="
    margin-top:14px;
    margin-bottom:0;
    font-size:14px;
    line-height:1.7;
    color:#111827;
">
<strong>Conclusión técnica:</strong> Se calcularon promedios diarios por comuna y estación, aplicando el criterio de representatividad mínima de 18 registros horarios. Los episodios críticos fueron identificados mediante umbrales normativos para PM2.5 y PM10.
</p>
</div>
"""
display(HTML(html_eventos))

# -----------------------------------------------------------------------------
# 10) Evidencia controlada sin truncamiento
# -----------------------------------------------------------------------------

print("Vista controlada de promedios diarios:")
display(promedios_diarios.head(10))

if episodios_pm25 > 0:
    print("Primeros episodios críticos PM2.5:")
    display(mp25_critico.head(10))
else:
    print("✓ No se detectaron episodios críticos PM2.5 bajo el umbral definido.")

if episodios_pm10 > 0:
    print("Primeros episodios críticos PM10:")
    display(mp10_critico.head(10))
else:
    print("✓ No se detectaron episodios críticos PM10 bajo el umbral definido.")

# DataFrame final para etapas posteriores
df_eventos_criticos = promedios_diarios.copy()
```

```python
# =============================================================================
# 8. PANEL EXPLORATORIO: RELACIÓN PM2.5 Y VARIABLES METEOROLÓGICAS
# Visualización mediante Hexbin de densidad
# =============================================================================

import re
import unicodedata
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, HTML

# -----------------------------------------------------------------------------
# 1) Localización automática del DataFrame principal
# -----------------------------------------------------------------------------

candidatos_df = [
    "df",
    "df_limpio",
    "df_normalizado",
    "df_final",
    "df_procesado",
    "df_eventos",
    "df_eventos_criticos",
    "sinca_limpio",
    "dataset"
]

df_encontrado = None

for nombre in candidatos_df:
    if nombre in globals() and isinstance(globals()[nombre], pd.DataFrame):
        if len(globals()[nombre]) > 0:
            df_encontrado = nombre
            break

if df_encontrado is None:
    dataframes_memoria = [
        nombre for nombre, valor in globals().items()
        if isinstance(valor, pd.DataFrame)
    ]

    raise NameError(
        "No se encontró un DataFrame principal válido en memoria. "
        f"DataFrames disponibles: {dataframes_memoria}. "
        "Ejecute primero las etapas de carga, limpieza y normalización."
    )

print(f"DataFrame utilizado: {df_encontrado}")

df_plot = globals()[df_encontrado].copy()

# -----------------------------------------------------------------------------
# 2) Normalización segura de nombres de columnas
# -----------------------------------------------------------------------------

def normalizar_columna(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")

df_plot.columns = [normalizar_columna(c) for c in df_plot.columns]

# -----------------------------------------------------------------------------
# 3) Detección robusta de columnas
# -----------------------------------------------------------------------------

def detectar_columna(df_base, patrones):
    for col in df_base.columns:
        col_norm = normalizar_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None

variable_objetivo = detectar_columna(
    df_plot,
    ["mp2_5", "pm2_5", "mp25", "pm25"]
)

if variable_objetivo is None:
    raise KeyError(
        "No se detectó columna objetivo para PM2.5. "
        f"Columnas disponibles: {df_plot.columns.tolist()}"
    )

mapa_meteorologico = {
    "temperatura": ["temperatura", "temp"],
    "humedad": ["humedad", "humidity", "hr"],
    "presion": ["presion", "pressure"],
    "viento": ["viento", "velocidad_viento", "wind"],
    "radiacion": ["radiacion", "radiation", "solar"]
}

variables_disponibles = []
variables_faltantes = []

for nombre_estandar, patrones in mapa_meteorologico.items():
    col_detectada = detectar_columna(df_plot, patrones)

    if col_detectada is not None:
        if col_detectada not in variables_disponibles:
            variables_disponibles.append(col_detectada)
    else:
        variables_faltantes.append(nombre_estandar)

if not variables_disponibles:
    raise KeyError(
        "No se encontraron variables meteorológicas disponibles para graficar. "
        f"Columnas disponibles: {df_plot.columns.tolist()}"
    )

print(f"Variable objetivo detectada: {variable_objetivo}")
print(f"Variables meteorológicas detectadas: {variables_disponibles}")
print(f"Variables meteorológicas no detectadas: {variables_faltantes}")

# -----------------------------------------------------------------------------
# 4) Preparación de datos
# -----------------------------------------------------------------------------

columnas_analisis = [variable_objetivo] + variables_disponibles

for col in columnas_analisis:
    df_plot[col] = pd.to_numeric(df_plot[col], errors="coerce")

registros_antes = len(df_plot)

df_plot = (
    df_plot
    .dropna(subset=columnas_analisis)
    .reset_index(drop=True)
)

registros_finales = len(df_plot)
registros_eliminados = registros_antes - registros_finales

if registros_finales == 0:
    raise ValueError(
        "Después de eliminar nulos en PM2.5 y variables meteorológicas, "
        "no quedaron registros válidos para graficar."
    )

# -----------------------------------------------------------------------------
# 5) Correlaciones exploratorias
# -----------------------------------------------------------------------------

correlaciones = (
    df_plot[columnas_analisis]
    .corr(numeric_only=True)[variable_objetivo]
    .drop(variable_objetivo)
    .sort_values(key=lambda s: s.abs(), ascending=False)
    .reset_index()
)

correlaciones.columns = ["Variable meteorológica", "Correlación con PM2.5"]
correlaciones["Correlación con PM2.5"] = (
    correlaciones["Correlación con PM2.5"]
    .round(3)
)

variable_mas_relacionada = correlaciones.iloc[0]["Variable meteorológica"]
corr_mas_relacionada = correlaciones.iloc[0]["Correlación con PM2.5"]

# -----------------------------------------------------------------------------
# 6) Cuadro resumen HTML profesional
# -----------------------------------------------------------------------------

html_panel = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:20px;
    border-left:6px solid #1F3A5F;
    border-radius:10px;
    font-family:Arial,Helvetica,sans-serif;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;color:#1F3A5F;text-align:left;">
📈 8. Panel Exploratorio PM2.5 vs Variables Meteorológicas
</h3>
<table style="width:100%;border-collapse:collapse;font-size:14px;color:#111827;text-align:left;">
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">DataFrame utilizado</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{df_encontrado}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Variable objetivo detectada</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{variable_objetivo}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Tipo de visualización</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">Hexbin de densidad</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Variables meteorológicas disponibles</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{', '.join(variables_disponibles)}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Variables no disponibles</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{', '.join(variables_faltantes) if variables_faltantes else 'Ninguna'}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros iniciales</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_antes:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros usados en gráficos</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_finales:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Registros excluidos por nulos</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_eliminados:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Variable con mayor relación absoluta</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{variable_mas_relacionada}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Correlación observada</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{corr_mas_relacionada}</td></tr>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:14px;color:#111827;text-align:left;">
<strong>Conclusión técnica:</strong>
Se construyó un panel exploratorio mediante mapas de densidad tipo Hexbin,
técnica adecuada para datasets de gran volumen porque permite identificar zonas
de mayor concentración de observaciones sin saturar visualmente el gráfico.
</p>
</div>
"""
display(HTML(html_panel))

print("Correlaciones entre PM2.5 y variables meteorológicas:")
display(correlaciones)

# -----------------------------------------------------------------------------
# 7) Visualización Hexbin de densidad en tonos azules
# -----------------------------------------------------------------------------

n_vars = len(variables_disponibles)

fig, axes = plt.subplots(
    1,
    n_vars,
    figsize=(max(6 * n_vars, 12), 5),
    squeeze=False
)

axes = axes.flatten()

for i, var in enumerate(variables_disponibles):

    hb = axes[i].hexbin(
        df_plot[var],
        df_plot[variable_objetivo],
        gridsize=45,
        mincnt=1,
        cmap="Blues",
        bins="log"
    )

    titulo_var = var.replace("_", " ").capitalize()

    axes[i].set_title(
        f"PM2.5 vs {titulo_var}",
        fontsize=12,
        fontweight="bold"
    )

    axes[i].set_xlabel(titulo_var)
    axes[i].set_ylabel("PM2.5" if i == 0 else "")
    axes[i].grid(True, alpha=0.25)

    cbar = fig.colorbar(hb, ax=axes[i])
    cbar.set_label("Densidad logarítmica")

fig.suptitle(
    "Densidad exploratoria entre PM2.5 y variables meteorológicas",
    fontsize=16,
    fontweight="bold",
    y=1.05
)

plt.tight_layout()
plt.show()
```

```python
# =============================================================================
# 9. CONSTRUCCIÓN DEL DATAFRAME CON VARIABLES BOOLEANAS
# Proyecto F4 (Entrega final): Calidad del Aire en Santiago
# =============================================================================

import re
import unicodedata
import pandas as pd
from IPython.display import display, HTML

# -----------------------------------------------------------------------------
# 1) Funciones utilitarias
# -----------------------------------------------------------------------------

def normalizar_columna(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna(df_base, patrones):
    for col in df_base.columns:
        col_norm = normalizar_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


# -----------------------------------------------------------------------------
# 2) Localizar automáticamente DataFrame base
# -----------------------------------------------------------------------------

candidatos_df = [
    "df_filtrado",
    "df",
    "df_limpio",
    "df_normalizado",
    "df_final",
    "df_procesado",
    "sinca_limpio",
    "dataset"
]

df_base_nombre = None

for nombre in candidatos_df:
    if nombre in globals() and isinstance(globals()[nombre], pd.DataFrame):
        if len(globals()[nombre]) > 0:
            df_base_nombre = nombre
            break

if df_base_nombre is None:
    disponibles = [
        nombre for nombre, valor in globals().items()
        if isinstance(valor, pd.DataFrame)
    ]
    raise NameError(
        "No se encontró ningún DataFrame válido en memoria. "
        f"DataFrames disponibles: {disponibles}. "
        "Ejecute primero la carga, limpieza o normalización del dataset."
    )

df_bool = globals()[df_base_nombre].copy()
df_bool.columns = [normalizar_columna(c) for c in df_bool.columns]

print(f"DataFrame base utilizado: {df_base_nombre}")
print(f"Filas iniciales: {len(df_bool):,}")
print(f"Columnas iniciales: {len(df_bool.columns):,}")

# -----------------------------------------------------------------------------
# 3) Detección robusta de columnas principales
# -----------------------------------------------------------------------------

col_fecha = detectar_columna(df_bool, ["fecha"])
col_comuna = detectar_columna(df_bool, ["comuna"])
col_estacion = detectar_columna(df_bool, ["estacion"])
col_pm25 = detectar_columna(df_bool, ["mp2_5", "pm2_5", "mp25", "pm25"])

columnas_detectadas = {
    "fecha": col_fecha,
    "comuna": col_comuna,
    "estacion": col_estacion,
    "mp2_5": col_pm25
}

faltantes = [k for k, v in columnas_detectadas.items() if v is None]

if faltantes:
    raise KeyError(
        f"Faltan columnas críticas: {faltantes}. "
        f"Columnas disponibles: {df_bool.columns.tolist()}"
    )

df_bool = df_bool.rename(columns={
    col_fecha: "fecha",
    col_comuna: "comuna",
    col_estacion: "estacion",
    col_pm25: "mp2_5"
})

# -----------------------------------------------------------------------------
# 4) Crear solo_fecha
# -----------------------------------------------------------------------------

df_bool["fecha"] = pd.to_datetime(df_bool["fecha"], errors="coerce")
df_bool = df_bool.dropna(subset=["fecha"]).copy()
df_bool["solo_fecha"] = df_bool["fecha"].dt.date

# -----------------------------------------------------------------------------
# 5) Variables booleanas esperadas
# -----------------------------------------------------------------------------

variables_bool = [
    "inversion_termica",
    "es_finde",
    "es_festivo"
]

bool_presentes = [c for c in variables_bool if c in df_bool.columns]
bool_faltantes = [c for c in variables_bool if c not in df_bool.columns]

# Si no existen, se reconstruyen parcialmente
if "es_finde" not in df_bool.columns:
    df_bool["es_finde"] = df_bool["fecha"].dt.dayofweek.isin([5, 6]).astype(int)
    bool_presentes.append("es_finde")
    bool_faltantes = [c for c in bool_faltantes if c != "es_finde"]

if "es_festivo" not in df_bool.columns:
    df_bool["es_festivo"] = 0
    bool_presentes.append("es_festivo")
    bool_faltantes = [c for c in bool_faltantes if c != "es_festivo"]

if "inversion_termica" not in df_bool.columns:
    df_bool["inversion_termica"] = 0
    bool_presentes.append("inversion_termica")
    bool_faltantes = [c for c in bool_faltantes if c != "inversion_termica"]

bool_presentes = list(dict.fromkeys(bool_presentes))

# -----------------------------------------------------------------------------
# 6) Normalización de booleanas
# -----------------------------------------------------------------------------

for col in bool_presentes:
    df_bool[col] = (
        df_bool[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "sí": 1,
            "si": 1,
            "true": 1,
            "1": 1,
            "yes": 1,
            "no": 0,
            "false": 0,
            "0": 0
        })
        .fillna(0)
        .astype(int)
    )

# -----------------------------------------------------------------------------
# 7) Validación numérica PM2.5
# -----------------------------------------------------------------------------

df_bool["mp2_5"] = pd.to_numeric(df_bool["mp2_5"], errors="coerce")

registros_antes = len(df_bool)

df_bool = df_bool.dropna(
    subset=["mp2_5", "solo_fecha", "comuna", "estacion"]
).copy()

registros_despues = len(df_bool)
registros_eliminados = registros_antes - registros_despues

if registros_despues == 0:
    raise ValueError(
        "No quedaron registros válidos después de limpiar PM2.5, fecha, comuna y estación."
    )

# -----------------------------------------------------------------------------
# 8) Agregación diaria
# -----------------------------------------------------------------------------

columnas_grupo = ["solo_fecha", "comuna", "estacion"]

agg_dict = {
    "mp2_5": "mean"
}

for col in bool_presentes:
    agg_dict[col] = "first"

mp25_diario_con_boolean = (
    df_bool
    .groupby(columnas_grupo, as_index=False)
    .agg(agg_dict)
    .rename(columns={
        "mp2_5": "mp2_5_promedio_diario"
    })
)

# -----------------------------------------------------------------------------
# 9) Métricas ejecutivas
# -----------------------------------------------------------------------------

filas_finales = len(mp25_diario_con_boolean)
dias_analizados = mp25_diario_con_boolean["solo_fecha"].nunique()
comunas_analizadas = mp25_diario_con_boolean["comuna"].nunique()
estaciones_analizadas = mp25_diario_con_boolean["estacion"].nunique()

promedio_pm25 = mp25_diario_con_boolean["mp2_5_promedio_diario"].mean()
max_pm25 = mp25_diario_con_boolean["mp2_5_promedio_diario"].max()

# -----------------------------------------------------------------------------
# 10) Resumen ejecutivo HTML
# -----------------------------------------------------------------------------

html_bool = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:20px;
    border-left:6px solid #1F3A5F;
    border-radius:10px;
    font-family:Arial,Helvetica,sans-serif;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:22px;font-weight:700;color:#1F3A5F;text-align:left;">
🧩 9. Resultado de Construcción del DataFrame con Variables Booleanas
</h3>
<table style="width:100%;border-collapse:collapse;font-size:14px;color:#111827;text-align:left;">
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">DataFrame base utilizado</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{df_base_nombre}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Filas originales</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_antes:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Filas válidas tras limpieza</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_despues:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Filas eliminadas</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{registros_eliminados:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Filas finales agregadas</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{filas_finales:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Días analizados</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{dias_analizados:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Comunas analizadas</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{comunas_analizadas:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Estaciones analizadas</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{estaciones_analizadas:,}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Booleanas incorporadas</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{', '.join(bool_presentes)}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">Booleanas no disponibles originalmente</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{', '.join(bool_faltantes) if bool_faltantes else 'Ninguna'}</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">PM2.5 promedio diario global</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{promedio_pm25:.2f} µg/m³</td></tr>
  <tr><td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:left;">PM2.5 máximo diario</td><td style="padding:8px;border:1px solid #cbd5e1;text-align:left;">{max_pm25:.2f} µg/m³</td></tr>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:14px;color:#111827;text-align:left;">
<strong>Conclusión técnica:</strong>
Se reconstruyó un DataFrame diario por fecha, comuna y estación, conservando variables booleanas
relevantes para el análisis exploratorio. Esta estructura permite evaluar la distribución del PM2.5
según condiciones cualitativas como fin de semana, festivos e inversión térmica, fortaleciendo la
trazabilidad metodológica del análisis.
</p>
</div>
"""

display(HTML(html_bool))

# -----------------------------------------------------------------------------
# 11) Evidencia controlada
# -----------------------------------------------------------------------------

print("Vista preliminar del DataFrame diario con variables booleanas:")
display(mp25_diario_con_boolean.head(10))

print("Columnas finales:")
print(list(mp25_diario_con_boolean.columns))
```

```python
# =============================================================================
# 10 MATRICES DE CORRELACIÓN ORIGINALES POR COMUNA (DISEÑO VERTICAL INTEGRAL)
# Proyecto F4 (Entrega final): Calidad del Aire en Santiago
# =============================================================================

import re
import unicodedata
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML

# -----------------------------------------------------------------------------
# 1) Funciones utilitarias
# -----------------------------------------------------------------------------

def normalizar_columna(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna(df_base, patrones):
    for col in df_base.columns:
        col_norm = normalizar_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


# -----------------------------------------------------------------------------
# 2) Selección robusta del DataFrame base
# -----------------------------------------------------------------------------

candidatos_df = [
    "df",
    "df_limpio",
    "df_normalizado",
    "df_final",
    "df_procesado",
    "mp25_diario_con_boolean",
    "mp25_diario",
    "df_eventos_criticos",
    "promedios_diarios"
]

df_base_nombre = None

for nombre in candidatos_df:
    if nombre in globals() and isinstance(globals()[nombre], pd.DataFrame):
        if len(globals()[nombre]) > 0:
            df_base_nombre = nombre
            break

if df_base_nombre is None:
    disponibles = [n for n, v in globals().items() if isinstance(v, pd.DataFrame)]
    raise NameError(f"No se encontró un DataFrame válido. Disponibles: {disponibles}.")

df_corr_base = globals()[df_base_nombre].copy()

# Localizar la columna temporal
col_fecha = detectar_columna(df_corr_base, ["fecha", "date", "time", "timestamp"])
if col_fecha:
    df_corr_base[col_fecha] = pd.to_datetime(df_corr_base[col_fecha])
    df_corr_base['fecha_dia'] = df_corr_base[col_fecha].dt.date
elif isinstance(df_corr_base.index, pd.DatetimeIndex):
    df_corr_base['fecha_dia'] = df_corr_base.index.date
else:
    raise KeyError("No se encontró columna temporal para realizar agregaciones diarias.")

# Localizar columna espacial
col_comuna_orig = detectar_columna(df_corr_base, ["comuna", "station", "estacion", "comunas"])
if not col_comuna_orig:
    raise KeyError("No se encontró una columna que represente las comunas.")

comunas_series = df_corr_base[col_comuna_orig].copy()
df_corr_base.columns = [normalizar_columna(c) for c in df_corr_base.columns]
col_comuna = normalizar_columna(col_comuna_orig)
df_corr_base[col_comuna] = comunas_series

# -----------------------------------------------------------------------------
# 3) Detección robusta de variables y mapeo de unidades de medida
# -----------------------------------------------------------------------------

mapa_variables = {
    "mp2_5": ["mp2_5", "pm2_5", "mp25", "pm25", "mp2_5_ug_m3", "mp2_5_promedio_diario"],
    "temperatura": ["temperatura", "temp"],
    "humedad": ["humedad", "humidity", "hr"],
    "presion": ["presion", "pressure"],
    "viento": ["viento", "velocidad_viento", "wind"],
    "radiacion": ["radiacion", "radiation", "solar"]
}

unidades_medida = {
    "mp2_5": "µg/m³",
    "temperatura": "°C",
    "humedad": "%",
    "presion": "hPa",
    "viento": "m/s",
    "radiacion": "W/m²"
}

columnas_detectadas = {}
variables_faltantes = []

for nombre_estandar, patrones in mapa_variables.items():
    col = detectar_columna(df_corr_base, patrones)
    if col is not None:
        columnas_detectadas[nombre_estandar] = col
    else:
        variables_faltantes.append(f"{nombre_estandar} ({unidades_medida[nombre_estandar]})")

if "mp2_5" not in columnas_detectadas:
    raise KeyError(f"No se detectó la columna objetivo de MP2.5.")

# -----------------------------------------------------------------------------
# 4) Preparación y Agregación Multivariada Diaria por Comuna
# -----------------------------------------------------------------------------

cols_analisis = [col_comuna, 'fecha_dia'] + list(columnas_detectadas.values())
df_corr = df_corr_base[cols_analisis].copy()

df_corr = df_corr.rename(columns={v: k for k, v in columnas_detectadas.items()})

for col in columnas_detectadas.keys():
    df_corr[col] = pd.to_numeric(df_corr[col], errors="coerce")

df_corr = df_corr.dropna().reset_index(drop=True)

if len(df_corr) == 0:
    raise ValueError("No quedaron registros válidos tras limpiar nulos.")

agg_dict = {"mp2_5": ["mean"]}
vars_meteorologicas = [v for v in columnas_detectadas.keys() if v != "mp2_5"]
for var in vars_meteorologicas:
    agg_dict[var] = ["min", "mean", "max"]

df_diario = df_corr.groupby([col_comuna, 'fecha_dia']).agg(agg_dict)

# Aplanar asignando etiquetas legibles con unidades
nuevas_cols = []
mapa_etiquetas_grafico = {} 

for col, estadistico in df_diario.columns:
    unidad = unidades_medida[col]
    if col == "mp2_5":
        nombre_col = "mp2_5"
        mapa_etiquetas_grafico[nombre_col] = f"mp2_5 ({unidad})"
        nuevas_cols.append(nombre_col)
    else:
        sufijo = "prom" if estadistico == "mean" else estadistico
        nombre_col = f"{col}_{sufijo}"
        mapa_etiquetas_grafico[nombre_col] = f"{col}_{sufijo} ({unidad})"
        nuevas_cols.append(nombre_col)

df_diario.columns = nuevas_cols
df_diario = df_diario.reset_index()

lista_comunas = sorted(df_diario[col_comuna].unique())
total_comunas = len(lista_comunas)

# Columnas definitivas del mapa de calor cruzado
columnas_finales_corr = ["mp2_5"] + [f"{v}_{m}" for v in vars_meteorologicas for m in ["min", "prom", "max"]]
etiquetas_ejes_con_unidades = [mapa_etiquetas_grafico[c] for c in columnas_finales_corr]

# -----------------------------------------------------------------------------
# 5) Procesamiento de datos y Construcción de Reporte HTML con Variable Mayor
# -----------------------------------------------------------------------------

filas_html_comunas = ""

for comuna_nom in lista_comunas:
    df_sub = df_diario[df_diario[col_comuna] == comuna_nom][columnas_finales_corr].copy()

    # Rellenar nulos locales con el promedio de la columna para evitar NaNs en la matriz
    df_sub = df_sub.fillna(df_sub.mean()).fillna(0)

    corr_comuna = df_sub.corr(numeric_only=True)

    # Capturar la fila de MP2.5 excluyendo la autocorrelación
    corr_pm25_comuna = corr_comuna["mp2_5"].drop("mp2_5")

    if len(corr_pm25_comuna.dropna()) > 0:
        # Encontrar la variable con mayor correlación absoluta
        var_mayor_col = corr_pm25_comuna.index[corr_pm25_comuna.abs().argmax()]
        valor_mayor_coef = corr_pm25_comuna[var_mayor_col]
        var_limpia_con_unidad = mapa_etiquetas_grafico[var_mayor_col]

        tipo_rel = "Positiva" if valor_mayor_coef > 0 else "Negativa" if valor_mayor_coef < 0 else "Nula"
        intensidad_rel = "Fuerte" if abs(valor_mayor_coef) >= 0.7 else "Moderada" if abs(valor_mayor_coef) >= 0.4 else "Débil"
    else:
        var_limpia_con_unidad = "Sin datos suficientes"
        valor_mayor_coef = 0.0
        tipo_rel = "N/A"
        intensidad_rel = ""

    filas_html_comunas += f"""
    <tr>
        <td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-transform:uppercase;">{comuna_nom}</td>
        <td style="padding:8px;border:1px solid #cbd5e1;color:#1F3A5F;font-weight:bold;">{var_limpia_con_unidad}</td>
        <td style="padding:8px;border:1px solid #cbd5e1;font-weight:bold;text-align:center;">{valor_mayor_coef:.3f}</td>
        <td style="padding:8px;border:1px solid #cbd5e1;font-style:italic;">{tipo_rel} {intensidad_rel}</td>
    </tr>
    """

# Renderizar panel HTML unificado
html_corr = f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#F4F7FA;
    color:#111827;
    padding:20px;
    border-left:6px solid #B22234;
    border-radius:10px;
    font-family:Arial,Helvetica,sans-serif;
    margin-top:16px;
    margin-bottom:16px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:22px;font-weight:700;color:#1F3A5F;text-align:left;">
🚨 10 Diagnóstico de Máxima Correlación Meteorológica por Comuna
</h3>
<table style="width:100%;border-collapse:collapse;font-size:13px;color:#111827;text-align:left;">
<thead>
    <tr style="background-color:#1F3A5F;color:#FFFFFF;">
        <th style="padding:10px;border:1px solid #cbd5e1;text-align:left;">Comuna Territorial</th>
        <th style="padding:10px;border:1px solid #cbd5e1;text-align:left;">Variable de Mayor Impacto (Unidad)</th>
        <th style="padding:10px;border:1px solid #cbd5e1;text-align:center;">Coeficiente (r)</th>
        <th style="padding:10px;border:1px solid #cbd5e1;text-align:left;">Interpretación Técnica</th>
    </tr>
</thead>
<tbody>
    {filas_html_comunas}
</tbody>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:14px;color:#111827;text-align:left;">
<strong>Conclusión del análisis espacial:</strong>
La tabla superior demuestra qué magnitud física ejerce el mayor control estadístico sobre las concentraciones de <strong>MP2.5</strong> en cada punto geográfico.
</p>
</div>
"""

display(HTML(html_corr))

# -----------------------------------------------------------------------------
# 6) Renderizado de Matrices Individuales (Un gráfico por fila/columna)
# -----------------------------------------------------------------------------

print(f"\nGenerando {total_comunas} gráficos independientes en disposición vertical...")

for comuna_nom in lista_comunas:
    df_sub = df_diario[df_diario[col_comuna] == comuna_nom][columnas_finales_corr].copy()
    
    # Tratamiento robusto anti-NaN por comuna
    df_sub = df_sub.fillna(df_sub.mean()).fillna(0)
    corr_comuna = df_sub.corr(numeric_only=True)
    
    # Inicializar cada gráfico con el tamaño original exacto (9, 7)
    fig, ax = plt.subplots(figsize=(9, 7))

    im = ax.imshow(
        corr_comuna.values,
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        aspect="auto"
    )

    ax.set_xticks(range(len(columnas_finales_corr)))
    ax.set_yticks(range(len(columnas_finales_corr)))

    ax.set_xticklabels(
        etiquetas_ejes_con_unidades,
        rotation=45,
        ha="right",
        fontsize=9
    )
    ax.set_yticklabels(etiquetas_ejes_con_unidades, fontsize=9)

    # Añadir valores numéricos celda por celda (Texto bicolor automático)
    for i in range(len(columnas_finales_corr)):
        for j in range(len(columnas_finales_corr)):
            valor = corr_comuna.values[i, j]
            texto_celda = f"{valor:.2f}" if not np.isnan(valor) else "0.00"
            
            ax.text(
                j, i, texto_celda,
                ha="center", va="center",
                color="white" if abs(valor) > 0.4 else "black",
                fontsize=9, fontweight="bold"
            )

    cb = fig.colorbar(im, ax=ax, shrink=0.8)
    cb.set_label("Coeficiente de correlación (r)", fontweight="bold")

    ax.set_title(
        f"MATRIZ DE CORRELACIÓN COMPLETA - COMUNA: {comuna_nom.upper()}",
        pad=14,
        fontweight="bold",
        fontsize=12
    )

    ax.grid(False)

    plt.tight_layout()
    plt.show()
    print(f"✅ Gráfico renderizado para: {comuna_nom.upper()}\n" + "-"*90)
```
