```python
# =============================================================================
# 2. CARGA Y PREPROCESAMIENTO MÍNIMO DEL DATASET
# Proyecto F4 (Entrega final) - Calidad del Aire Santiago
# =============================================================================

import os
import re
import unicodedata
import pandas as pd
import warnings
from IPython.display import HTML, display

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)

UMBRAL = 50

# -----------------------------------------------------------------------------
# Funciones utilitarias
# -----------------------------------------------------------------------------

def normalizar_columna(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"\s*\(.*?\)", "", col)
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna(df_base, patrones):
    for col in df_base.columns:
        col_norm = normalizar_columna(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


def cargar_datos(ruta: str) -> pd.DataFrame:
    if not os.path.exists(ruta):
        raise FileNotFoundError(f"No se encontró el archivo: {ruta}")

    with open(ruta, "r", encoding="latin-1") as f:
        primera_linea = f.readline()

    if ";" in primera_linea:
        df_local = pd.read_csv(ruta, sep=";", decimal=",", encoding="latin-1")
        formato_detectado = "CSV SINCA crudo / separador ; / decimal ,"
    else:
        df_local = pd.read_csv(ruta, encoding="utf-8")
        formato_detectado = "CSV procesado / separador estándar"

    df_local.columns = [normalizar_columna(c) for c in df_local.columns]

    return df_local, formato_detectado


def preprocesar(df_base: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    df_local = df_base.copy()

    registros_iniciales = len(df_local)
    columnas_iniciales = len(df_local.columns)

    col_fecha = detectar_columna(df_local, ["fecha"])
    col_pm25 = detectar_columna(df_local, ["mp2_5", "pm2_5", "mp25", "pm25", "mp2_5_ug_m3"])
    col_estacion = detectar_columna(df_local, ["estacion"])
    col_comuna = detectar_columna(df_local, ["comuna"])

    faltantes = []
    if col_fecha is None:
        faltantes.append("fecha")
    if col_pm25 is None:
        faltantes.append("mp2_5")

    if faltantes:
        raise KeyError(
            f"Faltan columnas críticas: {faltantes}. "
            f"Columnas disponibles: {df_local.columns.tolist()}"
        )

    rename_dict = {
        col_fecha: "fecha",
        col_pm25: "mp2_5"
    }

    if col_estacion is not None:
        rename_dict[col_estacion] = "estacion"

    if col_comuna is not None:
        rename_dict[col_comuna] = "comuna"

    df_local = df_local.rename(columns=rename_dict)

    df_local["fecha"] = pd.to_datetime(
        df_local["fecha"],
        errors="coerce",
        dayfirst=True
    )

    fechas_invalidas = int(df_local["fecha"].isna().sum())
    df_local = df_local.dropna(subset=["fecha"]).copy()

    variables_numericas = [
        "mp2_5",
        "mp10",
        "temperatura",
        "humedad",
        "presion",
        "viento",
        "radiacion"
    ]

    for col in variables_numericas:
        if col in df_local.columns:
            df_local[col] = pd.to_numeric(df_local[col], errors="coerce")

    nulos_pm25_antes = int(df_local["mp2_5"].isna().sum())

    if df_local["mp2_5"].isna().any():
        if "estacion" in df_local.columns:
            df_local["mp2_5"] = (
                df_local
                .groupby("estacion")["mp2_5"]
                .transform(lambda s: s.fillna(s.median()))
            )

        df_local["mp2_5"] = df_local["mp2_5"].fillna(df_local["mp2_5"].median())

    nulos_pm25_despues = int(df_local["mp2_5"].isna().sum())

    df_local["solo_fecha"] = df_local["fecha"].dt.date
    df_local["hora"] = df_local["fecha"].dt.hour
    df_local["mes"] = df_local["fecha"].dt.month
    df_local["anio_mes"] = df_local["fecha"].dt.to_period("M").astype(str)

    df_local = df_local.sort_values("fecha").reset_index(drop=True)

    df_local["critico"] = (df_local["mp2_5"] > UMBRAL).astype(int)

    metricas = {
        "registros_iniciales": registros_iniciales,
        "registros_finales": len(df_local),
        "registros_eliminados": registros_iniciales - len(df_local),
        "columnas_iniciales": columnas_iniciales,
        "columnas_finales": len(df_local.columns),
        "fecha_min": df_local["fecha"].min().strftime("%Y-%m-%d %H:%M:%S"),
        "fecha_max": df_local["fecha"].max().strftime("%Y-%m-%d %H:%M:%S"),
        "fechas_invalidas": fechas_invalidas,
        "nulos_pm25_antes": nulos_pm25_antes,
        "nulos_pm25_despues": nulos_pm25_despues,
        "horas_criticas": int(df_local["critico"].sum()),
        "porcentaje_critico": float(df_local["critico"].mean() * 100),
        "col_fecha_detectada": col_fecha,
        "col_pm25_detectada": col_pm25,
        "col_estacion_detectada": col_estacion if col_estacion else "No detectada",
        "col_comuna_detectada": col_comuna if col_comuna else "No detectada"
    }

    return df_local, metricas


# -----------------------------------------------------------------------------
# Ejecución
# -----------------------------------------------------------------------------

df_raw, formato_detectado = cargar_datos(str(RUTA_DATOS))
df, metricas_prepro = preprocesar(df_raw)

# -----------------------------------------------------------------------------
# Salida ejecutiva HTML
# -----------------------------------------------------------------------------

html_prepro = f"""
<div style="
width:100%;
max-width:100%;
box-sizing:border-box;
overflow-wrap:break-word;
background:#f8fafc;
border:1px solid #e2e8f0;
border-left:5px solid #1e40af;
padding:clamp(16px,3vw,22px);
border-radius:12px;
color:#1e293b;
font-family:Arial,sans-serif;
box-shadow:0 1px 3px rgba(0,0,0,.08);
text-align:left !important;
margin-top:14px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:clamp(20px,2.4vw,26px);font-weight:700;color:#1e40af;text-align:left !important;">
✅ Resultado de Carga y Preprocesamiento Mínimo
</h3>
<table style="
width:100%;
max-width:100%;
border-collapse:collapse;
font-size:clamp(13px,1.4vw,15px);
color:#1e293b;
text-align:left !important;
table-layout:fixed;
">
<tr><td style="width:38%;padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Ruta utilizada</td><td style="padding:8px;border:1px solid #e2e8f0;overflow-wrap:anywhere;text-align:left !important;">{RUTA_DATOS}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Formato detectado</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{formato_detectado}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Filas iniciales</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['registros_iniciales']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Filas finales</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['registros_finales']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Filas eliminadas</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['registros_eliminados']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columnas iniciales / finales</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['columnas_iniciales']} / {metricas_prepro['columnas_finales']}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Período disponible</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['fecha_min']} a {metricas_prepro['fecha_max']}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columna fecha detectada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['col_fecha_detectada']}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columna MP2.5 detectada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['col_pm25_detectada']}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Nulos MP2.5 antes / después</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['nulos_pm25_antes']:,} / {metricas_prepro['nulos_pm25_despues']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Umbral crítico MP2.5</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{UMBRAL} µg/m³</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Horas críticas detectadas</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['horas_criticas']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Porcentaje crítico</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_prepro['porcentaje_critico']:.2f}%</td></tr>
</table>

<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:clamp(13px,1.4vw,15px);color:#1e293b;text-align:left !important;">
<strong style="color:#1e40af;">Conclusión técnica:</strong>
El dataset fue cargado, normalizado y preprocesado con transformaciones mínimas orientadas al análisis algorítmico.
Se imputó MP2.5 mediante una estrategia robusta basada en mediana por estación y se construyó la variable objetivo
<strong style="color:#1e40af;">critico</strong>, preservando la mayor cantidad posible de observaciones para las etapas posteriores.
</p>
</div>
"""
display(HTML(html_prepro))

display(df.head())
```

```python
# =============================================================================
# 3. CODIFICACIÓN FUNCIONAL: DEL DATO AL EPISODIO
# Proyecto F4 (Entrega final) - Calidad del Aire Santiago
# =============================================================================

import re
import unicodedata
import numpy as np
import pandas as pd
from IPython.display import HTML, display

# -----------------------------------------------------------------------------
# Funciones utilitarias
# -----------------------------------------------------------------------------

def normalizar_columna_func(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna_func(df_base: pd.DataFrame, patrones: list[str]) -> str | None:
    for col in df_base.columns:
        col_norm = normalizar_columna_func(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


def serie_mp25(df_base: pd.DataFrame) -> tuple[np.ndarray, dict]:
    """
    Extrae la serie temporal de MP2.5 desde un DataFrame validado.

    Retorna:
        - np.ndarray con valores numéricos de MP2.5.
        - dict con métricas de control.
    """

    if not isinstance(df_base, pd.DataFrame):
        raise TypeError("El objeto recibido no es un DataFrame de pandas.")

    if df_base.empty:
        raise ValueError("El DataFrame recibido está vacío.")

    df_local = df_base.copy()
    df_local.columns = [normalizar_columna_func(c) for c in df_local.columns]

    col_pm25 = detectar_columna_func(
        df_local,
        ["mp2_5", "pm2_5", "mp25", "pm25", "mp2_5_ug_m3"]
    )

    if col_pm25 is None:
        raise KeyError(
            "No se detectó columna de MP2.5. "
            f"Columnas disponibles: {df_local.columns.tolist()}"
        )

    serie_raw = pd.to_numeric(df_local[col_pm25], errors="coerce")

    n_total = len(serie_raw)
    n_nulos = int(serie_raw.isna().sum())

    serie_limpia = serie_raw.dropna().to_numpy(dtype=float)

    if len(serie_limpia) == 0:
        raise ValueError("La serie MP2.5 quedó vacía después de eliminar nulos.")

    metricas = {
        "columna_detectada": col_pm25,
        "registros_originales": n_total,
        "nulos_excluidos": n_nulos,
        "horas_validas": len(serie_limpia),
        "mp25_min": float(np.min(serie_limpia)),
        "mp25_max": float(np.max(serie_limpia)),
        "mp25_promedio": float(np.mean(serie_limpia))
    }

    return serie_limpia, metricas


def detectar_episodios(serie: np.ndarray, umbral: float) -> list[tuple[int, int, int]]:
    """
    Detecta episodios críticos como rachas consecutivas sobre el umbral.

    Retorna:
        lista de tuplas: (indice_inicio, indice_fin, duracion_horas)
    """

    if not isinstance(serie, np.ndarray):
        raise TypeError("La serie debe ser un arreglo NumPy.")

    if serie.size == 0:
        raise ValueError("La serie MP2.5 está vacía.")

    if not np.issubdtype(serie.dtype, np.number):
        raise TypeError("La serie MP2.5 debe ser numérica.")

    if umbral <= 0:
        raise ValueError("El umbral debe ser un valor positivo.")

    episodios_locales = []
    inicio = None

    for i, valor in enumerate(serie):

        if valor > umbral and inicio is None:
            inicio = i

        elif valor <= umbral and inicio is not None:
            fin = i - 1
            duracion = fin - inicio + 1
            episodios_locales.append((inicio, fin, duracion))
            inicio = None

    if inicio is not None:
        fin = len(serie) - 1
        duracion = fin - inicio + 1
        episodios_locales.append((inicio, fin, duracion))

    return episodios_locales


# -----------------------------------------------------------------------------
# Ejecución del flujo funcional
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "No existe el DataFrame 'df'. Ejecute primero el punto 2 de carga y preprocesamiento."
    )

serie, metricas_serie = serie_mp25(df)
episodios = detectar_episodios(serie, UMBRAL)

duraciones = [e[2] for e in episodios]

episodios_detectados = len(episodios)
horas_criticas_en_rachas = int(sum(duraciones)) if duraciones else 0
duracion_minima = int(min(duraciones)) if duraciones else 0
duracion_maxima = int(max(duraciones)) if duraciones else 0
duracion_promedio = float(np.mean(duraciones)) if duraciones else 0
porcentaje_horas_criticas = (
    horas_criticas_en_rachas / len(serie) * 100
    if len(serie) > 0
    else 0
)

# -----------------------------------------------------------------------------
# Salida ejecutiva HTML
# -----------------------------------------------------------------------------

html_func = f"""
<div style="
width:100%;
max-width:100%;
box-sizing:border-box;
overflow-wrap:break-word;
background:#f8fafc;
border:1px solid #e2e8f0;
border-left:5px solid #1e40af;
padding:clamp(16px,3vw,22px);
border-radius:12px;
color:#1e293b;
font-family:Arial,sans-serif;
box-shadow:0 1px 3px rgba(0,0,0,.08);
text-align:left !important;
margin-top:14px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:clamp(20px,2.4vw,26px);font-weight:700;color:#1e40af;text-align:left !important;">
✅ Resultado de Codificación Funcional del Episodio
</h3>
<table style="
width:100%;
max-width:100%;
border-collapse:collapse;
font-size:clamp(13px,1.4vw,15px);
color:#1e293b;
text-align:left !important;
table-layout:fixed;
">
<tr><td style="width:38%;padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columna MP2.5 detectada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_serie['columna_detectada']}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Registros originales</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_serie['registros_originales']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Nulos excluidos</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_serie['nulos_excluidos']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Horas válidas en serie</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{metricas_serie['horas_validas']:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Umbral crítico utilizado</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{UMBRAL} µg/m³</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Episodios detectados</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{episodios_detectados:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Horas críticas acumuladas</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{horas_criticas_en_rachas:,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">% horas críticas</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{porcentaje_horas_criticas:.2f}%</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Duración mínima de episodio</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{duracion_minima} horas</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Duración máxima de episodio</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{duracion_maxima} horas</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Duración promedio</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{duracion_promedio:.2f} horas</td></tr>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:clamp(13px,1.4vw,15px);color:#1e293b;text-align:left !important;">
<strong style="color:#1e40af;">Conclusión técnica:</strong>
El flujo funcional transforma la serie MP2.5 en una estructura de episodios críticos basada en rachas consecutivas
sobre el umbral definido. Esta codificación entrega una base algorítmica clara, modular y reproducible para las
etapas posteriores de búsqueda, ordenamiento, complejidad y validación técnica.
</p>
</div>
"""
display(HTML(html_func))
```

```python
# =============================================================================
# 5. ALGORITMO RECURSIVO 2 — MERGE SORT DE HORAS POR MP2.5
# Proyecto F4 (Entrega final) - Calidad del Aire Santiago
# =============================================================================

import numpy as np
import pandas as pd
from IPython.display import HTML, display

# -----------------------------------------------------------------------------
# 1) Validaciones de entrada
# -----------------------------------------------------------------------------

if "serie" not in globals():
    raise NameError(
        "No existe la variable 'serie'. Ejecute primero el punto 3."
    )

serie_merge = np.asarray(serie, dtype=float)
serie_merge = serie_merge[~np.isnan(serie_merge)]

if serie_merge.size == 0:
    raise ValueError("La serie MP2.5 está vacía después de eliminar nulos.")

# Para demostrar recursividad sin sobrecargar el notebook
N_MUESTRA = min(1000, len(serie_merge))
muestra = serie_merge[:N_MUESTRA].tolist()

if len(muestra) == 0:
    raise ValueError("La muestra para Merge Sort quedó vacía.")

# -----------------------------------------------------------------------------
# 2) Implementación recursiva completa
# -----------------------------------------------------------------------------

def fusionar(izq: list[float], der: list[float]) -> list[float]:
    """
    Fusiona dos listas ordenadas en una única lista ordenada.
    """

    resultado = []
    i = 0
    j = 0

    while i < len(izq) and j < len(der):
        if izq[i] <= der[j]:
            resultado.append(izq[i])
            i += 1
        else:
            resultado.append(der[j])
            j += 1

    resultado.extend(izq[i:])
    resultado.extend(der[j:])

    return resultado


def merge_sort(arr: list[float]) -> list[float]:
    """
    Ordena una lista de forma recursiva usando Merge Sort.
    """

    if len(arr) <= 1:
        return arr

    medio = len(arr) // 2

    izquierda = merge_sort(arr[:medio])
    derecha = merge_sort(arr[medio:])

    return fusionar(izquierda, derecha)


# -----------------------------------------------------------------------------
# 3) Ejecución y validación
# -----------------------------------------------------------------------------

ordenada_merge = merge_sort(muestra)
ordenada_numpy = sorted(muestra)

validacion_ok = np.allclose(
    ordenada_merge,
    ordenada_numpy,
    equal_nan=False
)

primeros_5 = [round(x, 2) for x in ordenada_merge[:5]]
ultimos_5 = [round(x, 2) for x in ordenada_merge[-5:]]

minimo = float(np.min(ordenada_merge))
maximo = float(np.max(ordenada_merge))
promedio = float(np.mean(ordenada_merge))
mediana = float(np.median(ordenada_merge))

profundidad_teorica = int(np.ceil(np.log2(len(muestra)))) if len(muestra) > 0 else 0
complejidad = "O(n log n)"

# -----------------------------------------------------------------------------
# 4) Salida ejecutiva HTML
# -----------------------------------------------------------------------------

html_merge = f"""
<div style="
width:100%;
max-width:100%;
box-sizing:border-box;
overflow-wrap:break-word;
background:#f8fafc;
border:1px solid #e2e8f0;
border-left:5px solid #1e40af;
padding:clamp(16px,3vw,22px);
border-radius:12px;
color:#1e293b;
font-family:Arial,sans-serif;
box-shadow:0 1px 3px rgba(0,0,0,.08);
text-align:left !important;
margin-top:14px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:clamp(20px,2.4vw,26px);font-weight:700;color:#1e40af;text-align:left !important;">
✅ Resultado de Merge Sort Recursivo
</h3>
<table style="
width:100%;
max-width:100%;
border-collapse:collapse;
font-size:clamp(13px,1.4vw,15px);
color:#1e293b;
text-align:left !important;
table-layout:fixed;
">
<tr><td style="width:40%;padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Registros disponibles en serie</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{len(serie_merge):,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Tamaño de muestra ordenada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{len(muestra):,}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Primeros 5 valores ordenados</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{primeros_5}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Últimos 5 valores ordenados</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{ultimos_5}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Mínimo / mediana / promedio / máximo</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{minimo:.2f} / {mediana:.2f} / {promedio:.2f} / {maximo:.2f} µg/m³</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Validación contra sorted()</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{'✓ Correcto' if validacion_ok else '⚠ Revisar'}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Complejidad temporal</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{complejidad}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Profundidad recursiva estimada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{profundidad_teorica} niveles</td></tr>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:clamp(13px,1.4vw,15px);color:#1e293b;text-align:left !important;">
<strong style="color:#1e40af;">Conclusión técnica:</strong>
El algoritmo Merge Sort fue implementado correctamente mediante división recursiva y fusión ordenada.
El resultado ya no aparece vacío porque la función de fusión incorpora comparación, inserción y avance
de índices. La validación contra <strong style="color:#1e40af;">sorted()</strong> confirma la consistencia del ordenamiento.
</p>
</div>
"""
display(HTML(html_merge))
```

```python
# =============================================================================
# 7. IMPLEMENTACIÓN ORIENTADA A OBJETOS
# Encapsulamiento, herencia y polimorfismo
# =============================================================================

from abc import ABC, abstractmethod
import re
import unicodedata
import numpy as np
import pandas as pd
from IPython.display import HTML, display

# -----------------------------------------------------------------------------
# 1) Funciones utilitarias
# -----------------------------------------------------------------------------

def normalizar_columna_poo(col):
    col = str(col).strip()
    col = col.replace("µ", "u")
    col = unicodedata.normalize("NFKD", col)
    col = "".join(c for c in col if not unicodedata.combining(c))
    col = col.lower()
    col = re.sub(r"[^a-z0-9]+", "_", col)
    col = re.sub(r"_+", "_", col)
    return col.strip("_")


def detectar_columna_poo(df_base, patrones):
    for col in df_base.columns:
        col_norm = normalizar_columna_poo(col)
        if any(p in col_norm for p in patrones):
            return col
    return None


# -----------------------------------------------------------------------------
# 2) Clase base abstracta
# -----------------------------------------------------------------------------

class AnalizadorBase(ABC):
    """
    Clase base abstracta para el análisis de contaminantes atmosféricos.

    Principios aplicados:
    - Encapsulamiento: atributos internos protegidos.
    - Herencia: contrato común reutilizable.
    - Polimorfismo: cada subclase define su propio umbral crítico.
    """

    def __init__(self, df_base: pd.DataFrame, columna: str, nombre_contaminante: str):
        if not isinstance(df_base, pd.DataFrame):
            raise TypeError("df_base debe ser un DataFrame de pandas.")

        if df_base.empty:
            raise ValueError("df_base está vacío.")

        if columna not in df_base.columns:
            raise KeyError(
                f"La columna '{columna}' no existe en el DataFrame. "
                f"Columnas disponibles: {df_base.columns.tolist()}"
            )

        self._df = df_base.copy()
        self._columna = columna
        self._nombre_contaminante = nombre_contaminante

        self._serie = (
            pd.to_numeric(self._df[columna], errors="coerce")
            .dropna()
            .to_numpy(dtype=np.float64)
        )

        if len(self._serie) == 0:
            raise ValueError(
                f"La serie del contaminante {nombre_contaminante} quedó vacía después de eliminar nulos."
            )

    @property
    def columna(self) -> str:
        return self._columna

    @property
    def nombre_contaminante(self) -> str:
        return self._nombre_contaminante

    @property
    def serie(self) -> np.ndarray:
        return self._serie.copy()

    @property
    def total_registros(self) -> int:
        return len(self._serie)

    @abstractmethod
    def umbral_critico(self) -> float:
        pass

    def horas_criticas(self) -> int:
        return int(np.sum(self._serie > self.umbral_critico()))

    def porcentaje_critico(self) -> float:
        return self.horas_criticas() / self.total_registros * 100

    def resumen(self) -> dict:
        return {
            "contaminante": self._nombre_contaminante,
            "columna": self._columna,
            "umbral": self.umbral_critico(),
            "registros": self.total_registros,
            "media": round(float(np.mean(self._serie)), 2),
            "mediana": round(float(np.median(self._serie)), 2),
            "minimo": round(float(np.min(self._serie)), 2),
            "maximo": round(float(np.max(self._serie)), 2),
            "horas_criticas": self.horas_criticas(),
            "porcentaje_critico": round(self.porcentaje_critico(), 2)
        }


# -----------------------------------------------------------------------------
# 3) Subclases especializadas
# -----------------------------------------------------------------------------

class AnalizadorMP25(AnalizadorBase):
    """Analizador especializado para MP2.5."""

    def __init__(self, df_base: pd.DataFrame, columna: str, umbral_norma: float = 50.0):
        super().__init__(df_base, columna=columna, nombre_contaminante="MP2.5")
        self._umbral_norma = umbral_norma

    def umbral_critico(self) -> float:
        return self._umbral_norma

    def merge_sort_recursivo(self, arr: np.ndarray) -> np.ndarray:
        if len(arr) <= 1:
            return arr

        medio = len(arr) // 2
        izquierda = self.merge_sort_recursivo(arr[:medio])
        derecha = self.merge_sort_recursivo(arr[medio:])

        return self._merge(izquierda, derecha)

    def _merge(self, izq: np.ndarray, der: np.ndarray) -> np.ndarray:
        resultado = []
        i = 0
        j = 0

        while i < len(izq) and j < len(der):
            if izq[i] <= der[j]:
                resultado.append(izq[i])
                i += 1
            else:
                resultado.append(der[j])
                j += 1

        resultado.extend(izq[i:])
        resultado.extend(der[j:])

        return np.array(resultado, dtype=np.float64)

    def busqueda_binaria_recursiva(
        self,
        arr: np.ndarray,
        bajo: int,
        alto: int,
        valor: float
    ) -> int:
        if bajo > alto:
            return bajo if bajo < len(arr) else -1

        medio = (bajo + alto) // 2

        if arr[medio] >= valor:
            return self.busqueda_binaria_recursiva(arr, bajo, medio - 1, valor)

        return self.busqueda_binaria_recursiva(arr, medio + 1, alto, valor)


class AnalizadorMP10(AnalizadorBase):
    """Analizador especializado para MP10."""

    def __init__(self, df_base: pd.DataFrame, columna: str, umbral_norma: float = 150.0):
        super().__init__(df_base, columna=columna, nombre_contaminante="MP10")
        self._umbral_norma = umbral_norma

    def umbral_critico(self) -> float:
        return self._umbral_norma


# -----------------------------------------------------------------------------
# 4) Preparación robusta del DataFrame
# -----------------------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "No existe el DataFrame 'df'. Ejecute primero el punto 2."
    )

df_poo = df.copy()
df_poo.columns = [normalizar_columna_poo(c) for c in df_poo.columns]

col_pm25 = detectar_columna_poo(
    df_poo,
    ["mp2_5", "pm2_5", "mp25", "pm25", "mp2_5_ug_m3"]
)

col_pm10 = detectar_columna_poo(
    df_poo,
    ["mp10", "pm10", "mp10_ug_m3"]
)

if col_pm25 is None:
    raise KeyError(
        f"No se detectó columna MP2.5. Columnas disponibles: {df_poo.columns.tolist()}"
    )

analizadores = [
    AnalizadorMP25(df_poo, columna=col_pm25, umbral_norma=50.0)
]

if col_pm10 is not None:
    analizadores.append(
        AnalizadorMP10(df_poo, columna=col_pm10, umbral_norma=150.0)
    )

# -----------------------------------------------------------------------------
# 5) Ejecución polimórfica
# -----------------------------------------------------------------------------

resultados_poo = [analizador.resumen() for analizador in analizadores]
df_resultados_poo = pd.DataFrame(resultados_poo)

# Verificación técnica de métodos recursivos en MP2.5
analizador_mp25 = analizadores[0]
muestra_poo = analizador_mp25.serie[:min(1000, analizador_mp25.total_registros)]
muestra_ordenada_poo = analizador_mp25.merge_sort_recursivo(muestra_poo)

validacion_merge = np.allclose(
    muestra_ordenada_poo,
    np.sort(muestra_poo)
)

idx_binario_poo = analizador_mp25.busqueda_binaria_recursiva(
    muestra_ordenada_poo,
    0,
    len(muestra_ordenada_poo) - 1,
    analizador_mp25.umbral_critico()
)

# -----------------------------------------------------------------------------
# 6) Métricas ejecutivas
# -----------------------------------------------------------------------------

clases_implementadas = "AnalizadorBase, AnalizadorMP25"
if col_pm10 is not None:
    clases_implementadas += ", AnalizadorMP10"

polimorfismo_ok = len(analizadores) >= 2 if col_pm10 is not None else True

# -----------------------------------------------------------------------------
# 7) Salida ejecutiva HTML
# -----------------------------------------------------------------------------

html_poo = f"""
<div style="
width:100%;
max-width:100%;
box-sizing:border-box;
overflow-wrap:break-word;
background:#f8fafc;
border:1px solid #e2e8f0;
border-left:5px solid #1e40af;
padding:clamp(16px,3vw,22px);
border-radius:12px;
color:#1e293b;
font-family:Arial,sans-serif;
box-shadow:0 1px 3px rgba(0,0,0,.08);
text-align:left !important;
margin-top:14px;
">
<h3 style="margin-top:0;margin-bottom:15px;font-size:clamp(20px,2.4vw,26px);font-weight:700;color:#1e40af;text-align:left !important;">
✅ Resultado de Implementación Orientada a Objetos
</h3>
<table style="
width:100%;
max-width:100%;
border-collapse:collapse;
font-size:clamp(13px,1.4vw,15px);
color:#1e293b;
text-align:left !important;
table-layout:fixed;
">
<tr><td style="width:40%;padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Clases implementadas</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{clases_implementadas}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columna MP2.5 detectada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{col_pm25}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Columna MP10 detectada</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{col_pm10 if col_pm10 else 'No detectada'}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Encapsulamiento</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">✓ Atributos protegidos y propiedades de solo lectura</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Herencia</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">✓ Subclases heredan contrato de AnalizadorBase</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Polimorfismo</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{'✓ Métodos umbral_critico() especializados' if polimorfismo_ok else 'Parcial'}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Método plantilla</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">✓ resumen() reutiliza lógica común y delega umbral a subclases</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Validación Merge Sort MP2.5</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{'✓ Correcto' if validacion_merge else '⚠ Revisar'}</td></tr>
<tr><td style="padding:8px;border:1px solid #e2e8f0;background:#eff6ff;font-weight:bold;color:#1e3a8a;text-align:left !important;">Índice binario en muestra MP2.5</td><td style="padding:8px;border:1px solid #e2e8f0;text-align:left !important;">{idx_binario_poo if idx_binario_poo != -1 else 'Sin cruce'}</td></tr>
</table>
<p style="margin-top:16px;margin-bottom:0;line-height:1.6;font-size:clamp(13px,1.4vw,15px);color:#1e293b;text-align:left !important;">
<strong style="color:#1e40af;">Conclusión técnica:</strong>
La arquitectura orientada a objetos implementada cumple con encapsulamiento, herencia y polimorfismo.
La clase base define el contrato común y las subclases especializadas redefinen el umbral crítico,
permitiendo reutilizar el método de resumen sin modificar su lógica interna. Esto mejora cohesión,
mantenibilidad y extensibilidad del núcleo algorítmico.
</p>
</div>
"""
display(HTML(html_poo))
display(df_resultados_poo)
```
